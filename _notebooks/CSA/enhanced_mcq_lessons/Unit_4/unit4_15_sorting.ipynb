## Core Concept

The AP exam covers two sorting algorithms you need to be able to **trace and write**:

### Selection Sort
- Find the smallest element in the unsorted portion, swap it into position
- Outer loop: position being filled (left to right)
- Inner loop: find the minimum in the remaining unsorted section

### Insertion Sort
- Take each element and insert it into its correct position in the already-sorted left portion
- Like sorting a hand of playing cards: pick up one card at a time and slide it left until it fits
- Outer loop: current element being inserted
- Inner loop: shift elements right until the correct slot is found

| | Selection Sort | Insertion Sort |
|---|---|---|
| Strategy | Find min, swap into place | Insert each element into sorted left portion |
| Swaps | One per outer iteration | Shifts instead of swaps |
| Both are | O(n^2) -- nested loops | O(n^2) -- nested loops |
| Best for AP | Tracing is very common | Understanding shifts is key |

### Key Vocabulary

| Term | Meaning |
|---|---|
| **Selection sort** | Finds the minimum in the unsorted section; swaps it to the front |
| **Insertion sort** | Moves each element left until it is in the correct position |
| **In-place sort** | Rearranges elements within the original array |
| **O(n^2)** | Both algorithms have quadratic runtime -- slow on large data |

## Part 1 - Selection Sort


```java
double[] prices = {6.50, 4.50, 5.75, 6.00, 5.50};
String[] names  = {"Taro Boba","Oolong Tea","Matcha Latte","Classic Boba","Strawberry Matcha"};

System.out.println("Before: ");
for (int i = 0; i < prices.length; i++) System.out.printf("  $%.2f %s%n", prices[i], names[i]);

// Selection sort: sort by price ascending
for (int i = 0; i < prices.length - 1; i++) {
    // Find the index of the minimum in prices[i..end]
    int minIdx = i;
    for (int j = i + 1; j < prices.length; j++) {
        if (prices[j] < prices[minIdx]) {
            minIdx = j;
        }
    }
    // Swap prices[i] and prices[minIdx] -- and keep names in sync
    double tempPrice = prices[i];  prices[i]  = prices[minIdx];  prices[minIdx]  = tempPrice;
    String tempName  = names[i];   names[i]   = names[minIdx];   names[minIdx]   = tempName;

    System.out.println("Pass " + (i+1) + ": swapped " + names[i] + " into position " + i);
}

System.out.println("
Sorted by price:");
for (int i = 0; i < prices.length; i++) System.out.printf("  $%.2f  %s%n", prices[i], names[i]);
```

## Part 2 - Insertion Sort


```java
double[] prices = {6.50, 4.50, 5.75, 6.00, 5.50};
String[] names  = {"Taro Boba","Oolong Tea","Matcha Latte","Classic Boba","Strawberry Matcha"};

// Insertion sort: each item is slid left into the sorted portion
for (int i = 1; i < prices.length; i++) {
    double keyPrice = prices[i];
    String keyName  = names[i];
    int j = i - 1;

    // Shift elements right while they are larger than keyPrice
    while (j >= 0 && prices[j] > keyPrice) {
        prices[j + 1] = prices[j];
        names[j + 1]  = names[j];
        j--;
    }
    // Insert key into its correct slot
    prices[j + 1] = keyPrice;
    names[j + 1]  = keyName;

    System.out.print("After inserting " + keyName + ": ");
    for (double p : prices) System.out.printf("$%.2f ", p);
    System.out.println();
}

System.out.println("
Final sorted order:");
for (int i = 0; i < prices.length; i++) System.out.printf("  $%.2f  %s%n", prices[i], names[i]);
```

## Vocabulary Recap

| Term | Definition |
|---|---|
| **Selection sort** | Each pass finds the minimum in the unsorted section and swaps it to the front |
| **Insertion sort** | Each element is slid left into its correct position in the sorted left portion |
| **In-place** | The sort modifies the original array without creating a copy |
| **O(n^2)** | Both algorithms have quadratic runtime; acceptable for small arrays |
| **Pass** | One full execution of the outer loop body |
