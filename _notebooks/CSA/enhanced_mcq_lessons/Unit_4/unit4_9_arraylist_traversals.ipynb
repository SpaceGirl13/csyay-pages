## Core Concept

Traversing an ArrayList is similar to traversing an array -- but with key differences in syntax
and some dangerous traps when removing elements mid-loop.

### Three ways to traverse

```java
// 1. Standard for loop -- use when you need the index
for (int i = 0; i < list.size(); i++) { ... }

// 2. Enhanced for loop -- use for clean read-only traversal
for (String item : list) { ... }

// 3. While loop -- use for early exit or complex conditions
int i = 0;
while (i < list.size()) { ... i++; }
```

### The Big Trap: removing while looping forward

If you remove element at index `i`, every element after it shifts LEFT.
The next call to `list.get(i)` skips the element that just moved into that slot.

**Fix:** loop backwards, or adjust the index after a removal.

### Key Vocabulary

| Term | Meaning |
|---|---|
| **`size()`** | Returns the current number of elements -- call this fresh each iteration, not before the loop |
| **Index shifting** | Removing/inserting during a forward loop causes elements to slide, skipping or double-visiting |
| **Backward traversal** | Looping from `size()-1` down to 0; removals do not affect elements already visited |

## Part 1 - Standard and Enhanced For Loops


````java
import java.util.ArrayList;

ArrayList<String> orders = new ArrayList<String>();
orders.add("Matcha Latte");
orders.add("Taro Boba");
orders.add("Oolong Tea");
orders.add("Strawberry Matcha");
orders.add("Classic Boba");

// Standard for -- index visible, great for numbered receipts
System.out.println("=== Order Queue ===");
for (int i = 0; i < orders.size(); i++) {
    System.out.println("  #" + (i + 1) + "  " + orders.get(i));
}

System.out.println();

// Enhanced for -- clean and readable for simple reads
System.out.println("=== Drinks to make ===");
for (String drink : orders) {
    System.out.println("  >> " + drink);
}
````

## Part 2 - The Removal Trap (Read This Carefully)


````java
import java.util.ArrayList;

// Scenario: clear sold-out items from the menu
// Item is "sold out" if it contains "Boba" (running low today)

ArrayList<String> menu = new ArrayList<String>();
menu.add("Matcha Latte");      // 0
menu.add("Classic Boba");      // 1  <-- will be removed
menu.add("Oolong Tea");        // 2
menu.add("Taro Boba");         // 3  <-- will be removed
menu.add("Strawberry Matcha"); // 4

System.out.println("Before: " + menu);

// WRONG -- forward loop with removal skips elements
// After removing index 1 ("Classic Boba"), "Oolong Tea" slides to index 1
// The loop then jumps to index 2, never checking the new index 1.
// Do NOT do this -- shown here only so you recognize the bug.

// CORRECT -- loop backwards; removals only affect indices we've already passed
for (int i = menu.size() - 1; i >= 0; i--) {
    if (menu.get(i).contains("Boba")) {
        System.out.println("  Removing: " + menu.get(i));
        menu.remove(i);
    }
}

System.out.println("After:  " + menu);
````

````java
import java.util.ArrayList;

// Alternative fix: adjust the index when you remove while going forward
ArrayList<String> queue = new ArrayList<String>();
queue.add("Matcha Latte");
queue.add("Taro Boba");
queue.add("Oolong Tea");
queue.add("Classic Boba");
queue.add("Jasmine Tea");

System.out.println("Before: " + queue);

int i = 0;
while (i < queue.size()) {
    if (queue.get(i).contains("Boba")) {
        queue.remove(i);     // element at i is gone; next element slides into i
        // do NOT increment i -- re-check the same index
    } else {
        i++;                 // only advance if we did NOT remove
    }
}

System.out.println("After:  " + queue);
````

## Vocabulary Recap

| Term | Definition |
|---|---|
| **Forward traversal** | Loop from index 0 to size()-1; removal causes index shift bugs |
| **Backward traversal** | Loop from size()-1 to 0; safe for removal |
| **`size()`** | Current number of elements; evaluated fresh each iteration |
| **Index shifting** | Elements shift left after removal; can cause elements to be skipped |
| **ConcurrentModificationException** | Thrown when modifying a list during an enhanced for loop |
