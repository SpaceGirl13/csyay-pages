## Core Concept

Array algorithms are standard patterns you apply to a collection of data.
On the AP exam you are expected to both **trace** existing algorithms and **write** your own.

### The five essential patterns

| Algorithm | What it does |
|---|---|
| **Linear search** | Scan each element until a target is found |
| **Find max / min** | Track the best seen so far across all elements |
| **Compute sum / average** | Accumulate a running total |
| **Count** | Count elements that satisfy a condition |
| **Shift / reverse** | Rearrange elements within the array |

### Key Vocabulary

| Term | Meaning |
|---|---|
| **Linear search** | Checks each element one by one from index 0; O(n) time |
| **Sentinel value** | A special value (often -1 or null) returned when a search fails |
| **Accumulator** | A variable initialized before the loop that collects a running result |
| **In-place** | Modifying the array directly without creating a second array |
| **Temp variable** | A holder used during a swap: `temp = a; a = b; b = temp;` |

## Part 1 - Linear Search


````java
// Search for a drink by name; return its index or -1 if not found
public static int findDrink(String[] menu, String target) {
    for (int i = 0; i < menu.length; i++) {
        if (menu[i].equals(target)) {
            return i;           // found -- return the index immediately
        }
    }
    return -1;                  // sentinel: not found
}

String[] menu = {"Matcha Latte", "Classic Boba", "Oolong Tea", "Taro Boba", "Strawberry Matcha"};

int idx = findDrink(menu, "Oolong Tea");
System.out.println("Oolong Tea at index: " + idx);

int missing = findDrink(menu, "Lemon Soda");
System.out.println("Lemon Soda at index: " + missing);   // -1
````

````java
// Variation: search a double array for a price match
// Uses == for primitives (safe), NOT .equals()
public static int findPrice(double[] prices, double target) {
    for (int i = 0; i < prices.length; i++) {
        if (prices[i] == target) {
            return i;
        }
    }
    return -1;
}

double[] prices = {5.75, 6.00, 4.50, 6.50};
System.out.println("Index of $4.50: " + findPrice(prices, 4.50));
System.out.println("Index of $9.00: " + findPrice(prices, 9.00));
````

## Part 2 - Find Max and Min


````java
String[] menu   = {"Matcha Latte", "Classic Boba", "Oolong Tea", "Taro Boba"};
double[] prices = {5.75, 6.00, 4.50, 6.50};

// Initialize max and min to the FIRST element, then compare the rest
double maxPrice = prices[0];
int    maxIdx   = 0;
double minPrice = prices[0];
int    minIdx   = 0;

for (int i = 1; i < prices.length; i++) {   // start at 1
    if (prices[i] > maxPrice) {
        maxPrice = prices[i];
        maxIdx   = i;
    }
    if (prices[i] < minPrice) {
        minPrice = prices[i];
        minIdx   = i;
    }
}

System.out.println("Most expensive: " + menu[maxIdx]  + " ($" + maxPrice + ")");
System.out.println("Least expensive: " + menu[minIdx] + " ($" + minPrice + ")");
````

## Part 3 - Sum, Average, and Count


````java
int[] dailySales = {42, 38, 55, 49, 63, 71, 58};  // one week including weekend

// Sum
int total = 0;
for (int sales : dailySales) {
    total += sales;
}
System.out.println("Total drinks sold: " + total);

// Average
double avg = (double) total / dailySales.length;   // cast to avoid integer division
System.out.printf("Daily average:     %.1f drinks%n", avg);

// Count: how many days sold more than 50?
int highDays = 0;
for (int sales : dailySales) {
    if (sales > 50) {
        highDays++;
    }
}
System.out.println("Days above 50 sales: " + highDays);
````

## Part 4 - Reverse and Shift


````java
// Reverse an array in place using a two-pointer swap
String[] menu = {"Matcha Latte", "Classic Boba", "Oolong Tea", "Taro Boba"};

System.out.print("Before: ");
for (String d : menu) System.out.print(d + "  ");
System.out.println();

int left  = 0;
int right = menu.length - 1;
while (left < right) {
    String temp  = menu[left];   // temp holds left while we overwrite it
    menu[left]   = menu[right];
    menu[right]  = temp;
    left++;
    right--;
}

System.out.print("After:  ");
for (String d : menu) System.out.print(d + "  ");
System.out.println();
````

````java
// Shift all elements left by one; last slot set to default
int[] sales = {42, 38, 55, 49, 63};

System.out.print("Before shift: ");
for (int s : sales) System.out.print(s + " ");
System.out.println();

for (int i = 0; i < sales.length - 1; i++) {
    sales[i] = sales[i + 1];       // each element gets the value of its right neighbor
}
sales[sales.length - 1] = 0;       // last slot cleared

System.out.print("After shift:  ");
for (int s : sales) System.out.print(s + " ");
System.out.println();
````

## Vocabulary Recap

| Term | Definition |
|---|---|
| **Linear search** | Scans each element from index 0 until the target is found or the array ends |
| **Sentinel value** | Special return value (usually -1 or null) indicating a search found nothing |
| **Accumulator** | Variable initialized before the loop; updated each iteration to build a result |
| **In-place** | Algorithm that modifies the original array directly without a copy |
| **Temp variable** | Temporary holder used during a swap to prevent data loss |
