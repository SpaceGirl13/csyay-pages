## Core Concept

**Traversal** means visiting every element of an array, one at a time, to read or process it.

### Three traversal patterns in Java

**1. Standard for loop** - use when you need the index
```java
for (int i = 0; i < arr.length; i++) {
    System.out.println(arr[i]);
}
```

**2. Enhanced for loop (for-each)** - cleaner when you only need the value
```java
for (String item : arr) {
    System.out.println(item);
}
```

**3. while loop** - use when the stopping condition is not simply the end of the array
```java
int i = 0;
while (i < arr.length) {
    System.out.println(arr[i]);
    i++;
}
```

### When to use each

| Pattern | Use when |
|---|---|
| Standard `for` | You need the index `i` (to modify, compare neighbors, skip elements) |
| Enhanced `for` | Read-only traversal; you only need the value |
| `while` | Stopping condition is complex or not known at the start |

### Key Vocabulary

| Term | Meaning |
|---|---|
| **Traversal** | Visiting each element of an array in sequence |
| **Standard for loop** | `for (int i = 0; i < arr.length; i++)` -- index-based |
| **Enhanced for loop** | `for (Type var : arr)` -- value-based, read-only |
| **Accumulator** | A variable that builds up a result across iterations (sum, count, etc.) |
| **Off-by-one error** | Bug where a loop runs one time too many or too few |

## Part 1 - Standard For Loop Traversal


````java
// Print every drink on the menu with its position number
String[] menu = {"Matcha Latte", "Classic Boba", "Oolong Tea", "Taro Boba", "Strawberry Matcha"};

for (int i = 0; i < menu.length; i++) {
    System.out.println((i + 1) + ". " + menu[i]);   // display 1-based for readability
}
````

````java
// Accumulator pattern: sum all prices
double[] prices = {5.75, 6.00, 4.50, 6.50, 5.50};
double total = 0.0;                // accumulator starts at 0

for (int i = 0; i < prices.length; i++) {
    total += prices[i];            // add each element to the running total
}

System.out.printf("Total of all prices: $%.2f%n", total);
System.out.printf("Average price:       $%.2f%n", total / prices.length);
````

## Part 2 - Enhanced For Loop (for-each)


````java
// Enhanced for: cleaner syntax when the index is not needed
String[] menu = {"Matcha Latte", "Classic Boba", "Oolong Tea", "Taro Boba"};

for (String drink : menu) {
    System.out.println(drink);    // 'drink' holds the current element's value
}
````

````java
// Count how many items cost less than $6.00
double[] prices = {5.75, 6.00, 4.50, 6.50, 5.50};
int count = 0;

for (double price : prices) {    // 'price' is a copy of each element
    if (price < 6.00) {
        count++;
    }
}

System.out.println("Items under $6.00: " + count);

// IMPORTANT: modifying 'price' inside the loop does NOT change the array
// Enhanced for is read-only -- use standard for loop to modify elements
````

## Part 3 - While Loop Traversal


````java
// While loop: useful when you want to stop early
// Find the first drink whose name starts with 'T'
String[] menu = {"Matcha Latte", "Classic Boba", "Oolong Tea", "Taro Boba"};
int i = 0;
String found = "None";

while (i < menu.length) {
    if (menu[i].charAt(0) == 'T') {
        found = menu[i];
        break;          // stop as soon as we find one
    }
    i++;
}

System.out.println("First item starting with T: " + found);
System.out.println("Found at index: " + i);
````

## Part 4 - Finding Max and Min (Very Common MCQ Pattern)


````java
// Find the most and least expensive drinks
String[] menu   = {"Matcha Latte", "Classic Boba", "Oolong Tea", "Taro Boba"};
double[] prices = {5.75, 6.00, 4.50, 6.50};

// Start with index 0 as the current best -- NOT 0.0 as the value
double maxPrice = prices[0];
String maxName  = menu[0];
double minPrice = prices[0];
String minName  = menu[0];

for (int i = 1; i < prices.length; i++) {   // start at 1 -- already stored index 0
    if (prices[i] > maxPrice) {
        maxPrice = prices[i];
        maxName  = menu[i];
    }
    if (prices[i] < minPrice) {
        minPrice = prices[i];
        minName  = menu[i];
    }
}

System.out.println("Most expensive:  " + maxName + " ($" + maxPrice + ")");
System.out.println("Least expensive: " + minName + " ($" + minPrice + ")");
````

## Part 5 - Modifying Array Elements (Requires Standard For)


````java
// Apply a 10% discount to all prices -- must use standard for to modify
double[] prices = {5.75, 6.00, 4.50, 6.50};

System.out.println("Before discount:");
for (double p : prices) { System.out.printf("  $%.2f%n", p); }

// Modify in place -- enhanced for cannot do this
for (int i = 0; i < prices.length; i++) {
    prices[i] = prices[i] * 0.90;    // 10% off
}

System.out.println("After 10% discount:");
for (double p : prices) { System.out.printf("  $%.2f%n", p); }
````

## Vocabulary Recap

| Term | Definition |
|---|---|
| **Traversal** | Processing each element of an array in sequence |
| **Standard for loop** | Index-based; needed when you must modify elements or use the index |
| **Enhanced for loop** | Value-based; cleaner for read-only traversals; cannot modify array elements |
| **while loop traversal** | Useful for early termination or complex stopping conditions |
| **Accumulator** | Variable initialized before the loop; updated each iteration to build a result |
| **Off-by-one error** | Loop runs one too many or too few times; common with <= vs < |
