## Searching Algorithms

> Linear search checks everything. Binary search is smarter -- but needs a sorted list.

---


## Core Concept

Two fundamental ways to find something in an array:

### Linear Search
- Check every element one by one, starting at index 0
- Works on **any** array -- sorted or not
- Worst case: n comparisons (look at everything)

### Binary Search
- Only works on a **sorted** array
- Each comparison eliminates **half** the remaining elements
- Worst case: log2(n) comparisons -- dramatically faster on large data sets

| | Linear Search | Binary Search |
|---|---|---|
| Array must be sorted? | No | Yes |
| Best case | O(1) -- found at index 0 | O(1) -- found in the middle |
| Worst case | O(n) -- check everything | O(log n) -- halve each time |
| Code complexity | Simple | Moderate |

### Key Vocabulary

| Term | Meaning |
|---|---|
| **Linear search** | Scan from index 0; check each element; return index or -1 |
| **Binary search** | Repeatedly halve the search range using `low`, `high`, `mid` pointers |
| **`low`** | Left boundary of the current search range |
| **`high`** | Right boundary of the current search range |
| **`mid`** | Middle index: `(low + high) / 2` |
| **O(n) / O(log n)** | Informal runtime: linear vs. logarithmic growth |

## Part 1 - Linear Search


```java
// Linear search: find a drink price by name
String[] menu   = {"Classic Boba","Jasmine Tea","Matcha Latte","Oolong Tea","Taro Boba"};
double[] prices = {6.00, 4.00, 5.75, 4.50, 6.50};

public static int linearSearch(String[] arr, String target) {
    for (int i = 0; i < arr.length; i++) {
        System.out.println("  Checking [" + i + "]: " + arr[i]);   // show work
        if (arr[i].equals(target)) return i;
    }
    return -1;
}

System.out.println("Searching for 'Oolong Tea':");
int result = linearSearch(menu, "Oolong Tea");
System.out.println("Found at index: " + result + "  Price: $" + prices[result]);

System.out.println();
System.out.println("Searching for 'Espresso':");
System.out.println("Result: " + linearSearch(menu, "Espresso"));
```

## Part 2 - Binary Search


```java
// Binary search: the menu must be sorted alphabetically first
// (This one is already sorted)
String[] menu = {"Classic Boba","Jasmine Tea","Matcha Latte","Oolong Tea","Taro Boba"};

public static int binarySearch(String[] arr, String target) {
    int low  = 0;
    int high = arr.length - 1;

    while (low <= high) {
        int mid = (low + high) / 2;
        int cmp = target.compareTo(arr[mid]);   // negative if target < arr[mid]

        System.out.println("  low=" + low + " mid=" + mid + " high=" + high
                           + "  checking: " + arr[mid]);

        if      (cmp == 0) return mid;     // found!
        else if (cmp  < 0) high = mid - 1; // target is to the LEFT -- discard right half
        else               low  = mid + 1; // target is to the RIGHT -- discard left half
    }
    return -1;
}

System.out.println("Binary searching for 'Oolong Tea':");
int idx = binarySearch(menu, "Oolong Tea");
System.out.println("Found at index: " + idx);

System.out.println();
System.out.println("Binary searching for 'Matcha Latte':");
System.out.println("Found at index: " + binarySearch(menu, "Matcha Latte"));
```

## Part 3 - Why Binary Search Is Faster


```java
// Compare how many steps each takes on a large sorted list
// (Simulated -- not actually searching 1000 items)

int n = 1000;
System.out.println("List size: " + n);
System.out.println("Linear search worst case:  " + n + " comparisons");

int binarySteps = 0;
int remaining   = n;
while (remaining > 1) {
    remaining = remaining / 2;
    binarySteps++;
}
System.out.println("Binary search worst case:  ~" + binarySteps + " comparisons");

// Try 1,000,000 items
n = 1_000_000;
System.out.println();
System.out.println("List size: " + n);
System.out.println("Linear worst case:  " + n + " comparisons");
binarySteps = 0; remaining = n;
while (remaining > 1) { remaining /= 2; binarySteps++; }
System.out.println("Binary worst case:  ~" + binarySteps + " comparisons");
```

## Vocabulary Recap

| Term | Definition |
|---|---|
| **Linear search** | Checks each element in order; O(n) worst case |
| **Binary search** | Requires sorted array; halves the range each step; O(log n) worst case |
| **`low` / `high`** | Left and right boundaries of the current search window |
| **`mid`** | `(low + high) / 2`; the element compared each iteration |
| **O(n) vs O(log n)** | Linear grows with n; logarithmic grows much slower |
