## 2D Array Creation and Access

> The cafe tracks daily sales for every drink. That's a table. Tables are 2D arrays.

---


## Core Concept

A **2D array** is an array of arrays -- picture a grid with rows and columns.

```java
int[][] grid = new int[rows][cols];
grid[row][col]   // access one cell
```

Think of it as a cafe weekly sales board:
- Each **row** is a day of the week
- Each **column** is a drink

|  | Matcha | Boba | Oolong | Taro |
|---|---|---|---|---|
| **Mon** | 42 | 31 | 18 | 25 |
| **Tue** | 38 | 27 | 22 | 19 |
| **Wed** | 55 | 40 | 15 | 30 |

`sales[0][0]` = Monday Matcha = 42
`sales[2][3]` = Wednesday Taro = 30

### Key Vocabulary

| Term | Meaning |
|---|---|
| **2D array** | Array of arrays; accessed with two indices `[row][col]` |
| **Row** | First index; think horizontal slice |
| **Column** | Second index; think vertical slice |
| **`arr.length`** | Number of rows |
| **`arr[0].length`** | Number of columns (length of one row) |
| **Row-major order** | Java stores 2D arrays row by row in memory |

## Part 1 - Creating a 2D Array


````java
// The weekly cafe sales board
// Rows = days (0=Mon, 1=Tue, 2=Wed, 3=Thu, 4=Fri)
// Cols = drinks (0=Matcha, 1=Boba, 2=Oolong, 3=Taro)

int[][] sales = {
    {42, 31, 18, 25},   // Monday
    {38, 27, 22, 19},   // Tuesday
    {55, 40, 15, 30},   // Wednesday
    {49, 35, 20, 28},   // Thursday
    {63, 52, 25, 41}    // Friday
};

String[] days   = {"Mon", "Tue", "Wed", "Thu", "Fri"};
String[] drinks = {"Matcha", "Boba", "Oolong", "Taro"};

System.out.println("Rows (days):    " + sales.length);
System.out.println("Cols (drinks):  " + sales[0].length);

// Access specific cells
System.out.println("Monday Matcha:    " + sales[0][0]);
System.out.println("Wednesday Taro:   " + sales[2][3]);
System.out.println("Friday Boba:      " + sales[4][1]);
System.out.println("Last cell:        " + sales[sales.length-1][sales[0].length-1]);
````

````java
// Creating with new (values default to 0)
int rows = 5, cols = 4;
int[][] board = new int[rows][cols];

// Fill a specific cell
board[2][1] = 99;   // Wednesday Boba = 99

System.out.println("board[2][1] = " + board[2][1]);
System.out.println("board[0][0] = " + board[0][0]);   // still 0 -- default
````

## Part 2 - Row and Column Access


````java
int[][] sales = {
    {42, 31, 18, 25},
    {38, 27, 22, 19},
    {55, 40, 15, 30},
    {49, 35, 20, 28},
    {63, 52, 25, 41}
};

// Get all sales for Wednesday (row 2)
System.out.print("Wednesday sales: ");
for (int col = 0; col < sales[2].length; col++) {
    System.out.print(sales[2][col] + "  ");
}
System.out.println();

// Get all Matcha sales (column 0) across the week
System.out.print("Matcha all week: ");
for (int row = 0; row < sales.length; row++) {
    System.out.print(sales[row][0] + "  ");
}
System.out.println();
````

## Vocabulary Recap

| Term | Definition |
|---|---|
| **2D array** | An array of arrays; `type[][] name = new type[rows][cols]` |
| **`arr.length`** | Number of rows in the 2D array |
| **`arr[r].length`** | Number of columns in row r (number of columns) |
| **`arr[r][c]`** | Element at row r, column c |
| **Row-major** | Java stores 2D arrays as an array of row arrays |
