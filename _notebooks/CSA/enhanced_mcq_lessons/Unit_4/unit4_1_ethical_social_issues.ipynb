## Core Concept

Programs collect and store data about users and their behavior.
This creates serious ethical responsibilities -- just because you *can* collect data doesn't mean you *should*.

The AP exam tests your ability to reason about these tradeoffs, not just write code.

### What kind of data does a cafe app collect?

| Data type | Example | Sensitivity |
|---|---|---|
| Order history | "Matcha Latte, 3x per week" | Medium |
| Location data | "Usually orders from the Main St. location" | High |
| Payment info | Credit card number | Very high |
| Preferences | "No dairy, extra sugar" | Low-Medium |
| Browsing behavior | "Viewed but didn't buy the new seasonal item" | Medium |

### Key Vocabulary

| Term | Meaning |
|---|---|
| **Personally Identifiable Information (PII)** | Data that can identify a specific person (name, email, phone, location) |
| **Informed consent** | Users clearly understand and agree to what data is collected and how it is used |
| **Data privacy** | The right of individuals to control who accesses their personal information |
| **Data security** | Protecting stored data from unauthorized access or breaches |
| **Algorithmic bias** | When a program produces unfair outcomes for certain groups due to biased training data or design |
| **Least privilege** | Only collect and store the minimum data actually needed for the program to function |
| **Terms of service** | The legal agreement users accept that describes how their data is used |

```java
// A loyalty program stores customer data in variables and arrays.
// Consider what is necessary vs. what is invasive.

// Necessary for the loyalty program to function:
String customerName   = "Anishka";
String email          = "anishka@email.com";  // PII -- must be protected
int    loyaltyPoints  = 340;

// Useful but raises privacy questions:
double[] weeklySpend  = {12.50, 8.75, 15.00, 0.00, 9.25};  // 5 weeks of history
String   homeLocation = "37.7749,-122.4194";                 // GPS coordinates -- HIGH risk

// What happens if this data is leaked in a breach?
// What if it is sold to a third party without consent?

System.out.println("Customer: " + customerName);
System.out.println("Points:   " + loyaltyPoints);
System.out.println("Email stored (PII): " + email);
System.out.println("Location stored (sensitive): " + homeLocation);
```

## Part 2 - Algorithmic Bias

If a cafe app recommends drinks based on past orders, it may never suggest new items to someone with a short order history.
If it offers discounts based on spending patterns, lower-income customers may never qualify.

These are examples of **algorithmic bias** -- the program treats groups differently, often unintentionally.


```java
// A discount algorithm that unintentionally creates bias
public static String getDiscountTier(double totalSpentThisMonth) {
    if (totalSpentThisMonth >= 50.00) {
        return "Gold - 20% off";
    } else if (totalSpentThisMonth >= 25.00) {
        return "Silver - 10% off";
    } else {
        return "No discount";
    }
}

// A frequent customer who buys one small tea daily ($3.50 x 31 = $108.50)
System.out.println(getDiscountTier(108.50));  // Gold

// A customer who buys one large specialty drink weekly ($7.50 x 4 = $30)
System.out.println(getDiscountTier(30.00));   // Silver

// A customer who can only afford to visit twice a month ($5.00 x 2 = $10)
System.out.println(getDiscountTier(10.00));   // No discount
// Is this fair? The algorithm rewards frequency, which correlates with income.
```

## Part 3 - Designing Responsibly

Questions a developer should ask before collecting data:

1. **Do we need this?** If the app works without it, don't collect it.
2. **Have users consented?** Is the opt-in clear, or buried in fine print?
3. **How long do we keep it?** Delete data when it is no longer needed.
4. **Who has access?** Restrict access to the minimum number of people/systems.
5. **What if it leaks?** Design assuming a breach will happen eventually.


```java
// Safer design: store only what is needed, avoid raw PII where possible

// Instead of storing a real email, store a hashed identifier
// (In real Java you'd use a hashing library -- this illustrates the concept)
public static String anonymize(String email) {
    // Real hashing would use MessageDigest -- simplified here
    return "user_" + Math.abs(email.hashCode());
}

String rawEmail     = "anishka@email.com";
String anonymizedId = anonymize(rawEmail);

System.out.println("Raw email (risky to store): " + rawEmail);
System.out.println("Anonymized ID (safer):      " + anonymizedId);
// The program can still identify returning users without storing their real email
```

## Vocabulary Recap

| Term | Definition |
|---|---|
| **PII** | Data that identifies a specific person (name, email, location, etc.) |
| **Informed consent** | User clearly understands and agrees to data collection practices |
| **Data privacy** | Individual's right to control access to their personal information |
| **Data security** | Technical protections preventing unauthorized access to stored data |
| **Algorithmic bias** | Unintended unfair treatment of groups resulting from program design or training data |
