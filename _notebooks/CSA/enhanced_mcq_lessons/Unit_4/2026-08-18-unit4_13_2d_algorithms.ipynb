## Core Concept

2D array algorithms are the same patterns as 1D -- search, max/min, sum, count --
just applied across rows and columns.

The key skill is knowing **which dimension to fix** and **which to iterate**.

| Goal | Approach |
|---|---|
| Grand total | Sum every cell |
| Row total/average | Fix row, iterate columns |
| Column total/average | Fix column, iterate rows |
| Find max in whole grid | Track best seen across all rows and columns |
| Find max per row | Reset the max tracker at the start of each row |

## Part 1 - Row and Column Totals


````java
String[] days   = {"Mon","Tue","Wed","Thu","Fri"};
String[] drinks = {"Matcha","Boba","Oolong","Taro"};

int[][] sales = {
    {42, 31, 18, 25},
    {38, 27, 22, 19},
    {55, 40, 15, 30},
    {49, 35, 20, 28},
    {63, 52, 25, 41}
};

// Row totals: total drinks sold each day
System.out.println("Daily totals:");
for (int row = 0; row < sales.length; row++) {
    int dayTotal = 0;
    for (int col = 0; col < sales[row].length; col++) {
        dayTotal += sales[row][col];
    }
    System.out.println("  " + days[row] + ": " + dayTotal + " drinks");
}

System.out.println();

// Column totals: total sales per drink type
System.out.println("Drink totals:");
for (int col = 0; col < sales[0].length; col++) {
    int drinkTotal = 0;
    for (int row = 0; row < sales.length; row++) {
        drinkTotal += sales[row][col];
    }
    System.out.println("  " + drinks[col] + ": " + drinkTotal + " cups");
}
````

## Part 2 - Find the Global Max


````java
int[][] sales = {
    {42, 31, 18, 25},
    {38, 27, 22, 19},
    {55, 40, 15, 30},
    {49, 35, 20, 28},
    {63, 52, 25, 41}
};
String[] days   = {"Mon","Tue","Wed","Thu","Fri"};
String[] drinks = {"Matcha","Boba","Oolong","Taro"};

// Find the single best-selling drink-day combination
int maxSales = sales[0][0];
int maxRow = 0, maxCol = 0;

for (int row = 0; row < sales.length; row++) {
    for (int col = 0; col < sales[row].length; col++) {
        if (sales[row][col] > maxSales) {
            maxSales = sales[row][col];
            maxRow   = row;
            maxCol   = col;
        }
    }
}

System.out.println("Best combo: " + days[maxRow] + " " + drinks[maxCol]
                   + " with " + maxSales + " cups sold.");
````

## Part 3 - Count and Conditional Algorithms


````java
int[][] sales = {
    {42, 31, 18, 25},
    {38, 27, 22, 19},
    {55, 40, 15, 30},
    {49, 35, 20, 28},
    {63, 52, 25, 41}
};

// Count how many individual drink-day cells hit more than 40 cups
int highCount = 0;
for (int[] row : sales) {
    for (int val : row) {
        if (val > 40) highCount++;
    }
}
System.out.println("Cells above 40 cups: " + highCount);

// Which day had all drinks above 20?
String[] days = {"Mon","Tue","Wed","Thu","Fri"};
for (int row = 0; row < sales.length; row++) {
    boolean allAbove20 = true;
    for (int col = 0; col < sales[row].length; col++) {
        if (sales[row][col] <= 20) { allAbove20 = false; break; }
    }
    if (allAbove20) System.out.println(days[row] + " -- all drinks above 20!");
}
````

## Vocabulary Recap

| Term | Definition |
|---|---|
| **Row total** | Sum of all columns in a fixed row |
| **Column total** | Sum of all rows in a fixed column |
| **Global max** | The largest single value across the entire 2D array |
| **Main diagonal** | Elements where `row == col` in a square 2D array |
