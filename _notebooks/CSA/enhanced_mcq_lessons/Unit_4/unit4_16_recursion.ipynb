## Recursion

> A method that calls itself. Sounds circular. Turns out it's really useful.

---


## Core Concept

A **recursive method** calls itself with a smaller version of the problem,
until it reaches a **base case** that can be solved directly without another call.

Every recursive method needs two things:

1. **Base case** -- the condition that stops the recursion (no more calls)
2. **Recursive case** -- the call to itself with a smaller/simpler input

```
countdown(3)
  -> countdown(2)
       -> countdown(1)
            -> countdown(0)  // BASE CASE -- stop here
```

Without a base case (or with a wrong one), recursion never stops -> **StackOverflowError**.

### Key Vocabulary

| Term | Meaning |
|---|---|
| **Recursive method** | A method that calls itself as part of its own body |
| **Base case** | The condition that stops recursion; returns a value directly |
| **Recursive case** | The part that calls the method again with a smaller input |
| **Call stack** | Each recursive call adds a stack frame; too many -> StackOverflowError |
| **StackOverflowError** | Thrown when recursion has no base case or never reaches it |

## Part 1 - A Simple Countdown


````java
// Count down from n and announce when the cafe opens
public static void openingCountdown(int n) {
    if (n == 0) {                               // BASE CASE
        System.out.println("We're open! Come get your boba!");
        return;
    }
    System.out.println(n + "...");              // RECURSIVE CASE
    openingCountdown(n - 1);                    // call with smaller n
}

openingCountdown(5);
````

## Part 2 - Recursive Math: Factorial


````java
// n! = n * (n-1) * (n-2) * ... * 1
// Cafe context: how many ways can 4 different drinks be arranged on a display shelf?

public static int factorial(int n) {
    if (n == 0 || n == 1) return 1;            // BASE CASE
    return n * factorial(n - 1);               // RECURSIVE CASE
}

// Trace for factorial(4):
// factorial(4) = 4 * factorial(3)
//                    3 * factorial(2)
//                        2 * factorial(1)
//                            = 1
//                        = 2
//                    = 6
//                = 24

System.out.println("4! = " + factorial(4));    // 24 arrangements of 4 drinks
System.out.println("5! = " + factorial(5));
System.out.println("1! = " + factorial(1));
System.out.println("0! = " + factorial(0));
````

## Part 3 - Recursive String Processing


````java
// Count how many times 'a' appears in a drink name -- recursively
public static int countA(String s) {
    if (s.length() == 0) return 0;             // BASE CASE: empty string has no 'a's

    // RECURSIVE CASE: check first character, recurse on the rest
    int firstIsA = (s.charAt(0) == 'a') ? 1 : 0;
    return firstIsA + countA(s.substring(1));  // substring(1) drops the first character
}

System.out.println("'Matcha Latte' has " + countA("Matcha Latte") + " a's");
System.out.println("'Taro Boba'    has " + countA("Taro Boba")    + " a's");
System.out.println("'Oolong Tea'   has " + countA("Oolong Tea")   + " a's");
````

## Part 4 - Tracing Recursion


````java
// Mystery method -- trace it before running to predict the output
public static int mystery(int n) {
    if (n <= 0) return 0;
    return n + mystery(n - 2);
}

// Trace mystery(6):
// mystery(6) = 6 + mystery(4)
//                  4 + mystery(2)
//                      2 + mystery(0)
//                          = 0
//                      = 2
//                  = 6
//              = 12

System.out.println("mystery(6) = " + mystery(6));
System.out.println("mystery(5) = " + mystery(5));   // predict first!
System.out.println("mystery(1) = " + mystery(1));
````
