## Core Concept

Every primitive type in Java has a corresponding **wrapper class** -- an object version of that type.
Wrapper classes are needed when a collection (like `ArrayList`) requires objects, not primitives.

### Primitive vs. Wrapper

| Primitive | Wrapper class |
|---|---|
| `int` | `Integer` |
| `double` | `Double` |
| `boolean` | `Boolean` |
| `char` | `Character` |

### Autoboxing and Unboxing

Java automatically converts between primitive and wrapper types:

- **Autoboxing** -- `int` -> `Integer` automatically (primitive to object)
- **Unboxing** -- `Integer` -> `int` automatically (object to primitive)

```java
Integer x = 5;          // autoboxing:   int 5 -> Integer
int     y = x;          // unboxing:     Integer -> int
```

### Key Vocabulary

| Term | Meaning |
|---|---|
| **Wrapper class** | An object that wraps a primitive value (`Integer`, `Double`, `Boolean`, `Character`) |
| **Autoboxing** | Automatic conversion from a primitive to its wrapper class |
| **Unboxing** | Automatic conversion from a wrapper object to its primitive |
| **`Integer.parseInt(s)`** | Converts a String to an int |
| **`Double.parseDouble(s)`** | Converts a String to a double |
| **`Integer.MAX_VALUE`** | Largest possible int value (2,147,483,647) |
| **`Integer.MIN_VALUE`** | Smallest possible int value (-2,147,483,648) |
| **`NullPointerException`** | Thrown when unboxing a null wrapper reference |

## Part 1 - Autoboxing and Unboxing


````java
// Autoboxing: Java wraps the primitive automatically
Integer calorieCount = 320;       // int 320 -> Integer object
Double  price        = 5.75;      // double -> Double object
Boolean isVegan      = true;      // boolean -> Boolean object

System.out.println("Calories: " + calorieCount);
System.out.println("Price:    " + price);
System.out.println("Vegan:    " + isVegan);

// Unboxing: Java unwraps the object automatically
int    cal    = calorieCount;     // Integer -> int
double p      = price;            // Double  -> double
boolean vegan = isVegan;          // Boolean -> boolean

System.out.println("Unboxed calories: " + cal);

// Arithmetic works directly on wrappers (unboxing happens behind the scenes)
Integer a = 10;
Integer b = 20;
int     sum = a + b;              // unbox both, add, result is primitive int
System.out.println("Sum: " + sum);
````

## Part 2 - Parsing Strings to Primitives


````java
// Reading data from a file or user input gives you Strings.
// Wrapper classes provide static parse methods to convert them.

String priceStr    = "6.50";
String quantityStr = "3";
String veganStr    = "true";

double  price    = Double.parseDouble(priceStr);      // "6.50"  -> 6.50
int     quantity = Integer.parseInt(quantityStr);     // "3"     -> 3
boolean isVegan  = Boolean.parseBoolean(veganStr);    // "true"  -> true

System.out.println("Price x quantity: $" + (price * quantity));
System.out.println("Is vegan: " + isVegan);

// Parsing a String that is not a valid number throws NumberFormatException
try {
    int bad = Integer.parseInt("Matcha");   // cannot convert a name to int
} catch (NumberFormatException e) {
    System.out.println("Caught: " + e.getMessage());
}
````

## Part 3 - Useful Constants and Methods


````java
// Integer.MAX_VALUE and MIN_VALUE are useful to initialize max/min searches
System.out.println("Integer.MAX_VALUE: " + Integer.MAX_VALUE);
System.out.println("Integer.MIN_VALUE: " + Integer.MIN_VALUE);

// Using MAX_VALUE to initialize a minimum-finder
int[]    dailySales = {42, 38, 55, 49, 63};
int      minSales   = Integer.MAX_VALUE;   // any real value will be smaller
int      minDay     = -1;

for (int i = 0; i < dailySales.length; i++) {
    if (dailySales[i] < minSales) {
        minSales = dailySales[i];
        minDay   = i;
    }
}
System.out.println("Slowest day index: " + minDay + "  (" + minSales + " drinks)");

// Converting a number to a String
int    loyalty  = 340;
String display  = Integer.toString(loyalty);
System.out.println("Points as String: "" + display + ""  length=" + display.length());
````

## Part 4 - null Wrappers and NullPointerException


````java
// A wrapper variable can be null -- a primitive cannot
Integer loyaltyPoints = null;   // valid: Integer is an object
// int x = null;               // compile error: primitive cannot be null

System.out.println("loyaltyPoints is null: " + (loyaltyPoints == null));

// Unboxing null throws NullPointerException
try {
    int pts = loyaltyPoints;    // attempts to unbox null -- crashes
    System.out.println(pts);
} catch (NullPointerException e) {
    System.out.println("Caught NullPointerException: cannot unbox null");
}
````

## Part 5 - Comparing Wrapper Objects


````java
// == compares references for objects, not values
// Use .equals() or .compareTo() for value comparison

Integer a = 200;
Integer b = 200;

System.out.println("a == b:        " + (a == b));          // may be false! (different objects)
System.out.println("a.equals(b):   " + a.equals(b));       // true  (same value)
System.out.println("compareTo:     " + a.compareTo(b));    // 0 means equal

// Note: Java caches Integer objects for values -128 to 127
// So == may return true for small numbers -- do NOT rely on this
Integer x = 50;
Integer y = 50;
System.out.println("x == y (small): " + (x == y));   // likely true, but bad practice
// Always use .equals() to compare Integer values safely
````
