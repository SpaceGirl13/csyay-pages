## Core Concept

An **array** is a fixed-size, ordered collection of values of the same type.
Once created, its size cannot change.

### Two ways to create an array

**1. Declare size, fill later:**
```java
double[] prices = new double[4];   // 4 slots, all initialized to 0.0
```

**2. Initializer list (size set automatically):**
```java
double[] prices = {5.75, 6.00, 4.50, 6.50};
```

### Accessing elements

```java
prices[0]              // first element (index 0)
prices[prices.length - 1]   // last element (always)
```

### Default values when declared with `new`

| Type | Default |
|---|---|
| `int` | 0 |
| `double` | 0.0 |
| `boolean` | false |
| `String` / any object | null |

### Key Vocabulary

| Term | Meaning |
|---|---|
| **Array** | Fixed-size, ordered collection of values of one type |
| **Index** | Integer position of an element; starts at 0 |
| **`length`** | Field (not method) that returns the number of elements in the array |
| **Initializer list** | `{v1, v2, v3}` syntax that creates and fills an array in one line |
| **`ArrayIndexOutOfBoundsException`** | Runtime error thrown when accessing an index < 0 or >= length |
| **Default value** | Value Java assigns to uninitialized array elements |

## Part 1 - Creating Arrays


````java
// Method 1: Declare size, fill elements manually
String[] drinkNames = new String[4];   // 4 slots, all null initially
drinkNames[0] = "Matcha Latte";
drinkNames[1] = "Classic Boba";
drinkNames[2] = "Oolong Tea";
drinkNames[3] = "Taro Boba";

System.out.println("First drink:  " + drinkNames[0]);
System.out.println("Last drink:   " + drinkNames[3]);
System.out.println("Array length: " + drinkNames.length);

System.out.println();

// Method 2: Initializer list -- cleaner when values are known upfront
double[] prices = {5.75, 6.00, 4.50, 6.50};
System.out.println("Cheapest price:    $" + prices[2]);
System.out.println("Most expensive:    $" + prices[prices.length - 1]);
````

## Part 2 - Accessing and Modifying Elements


````java
// Accessing: read a value at an index
int[] dailySales = {42, 38, 55, 49, 63};   // Mon-Fri

System.out.println("Monday sales:  " + dailySales[0]);
System.out.println("Friday sales:  " + dailySales[4]);
System.out.println("Last element:  " + dailySales[dailySales.length - 1]);

System.out.println();

// Modifying: assign a new value at an index
dailySales[2] = 60;   // Wednesday was recounted
System.out.println("Wednesday corrected: " + dailySales[2]);

// Arithmetic with elements
int total = dailySales[0] + dailySales[1] + dailySales[2];
System.out.println("Mon-Wed total: " + total);
````

## Part 3 - Default Values and null


````java
// int array: defaults to 0
int[] stockCount = new int[3];
System.out.println("Default int values: " + stockCount[0] + ", "
                                           + stockCount[1] + ", "
                                           + stockCount[2]);

// double array: defaults to 0.0
double[] revenue = new double[3];
System.out.println("Default double:     " + revenue[0]);

// boolean array: defaults to false
boolean[] inStock = new boolean[3];
System.out.println("Default boolean:    " + inStock[0]);

// String array: defaults to null
String[] names = new String[3];
System.out.println("Default String:     " + names[0]);

// Calling a method on a null element throws NullPointerException!
// System.out.println(names[0].length());  // would crash -- names[0] is null
````

## Part 4 - The ArrayIndexOutOfBoundsException (Common MCQ Trap)


````java
// Valid indices: 0 to array.length - 1
String[] drinks = {"Matcha", "Boba", "Oolong"};   // length = 3, valid: 0, 1, 2

System.out.println(drinks[0]);   // OK
System.out.println(drinks[2]);   // OK -- last valid index

// These would throw ArrayIndexOutOfBoundsException at runtime:
// System.out.println(drinks[3]);    // index 3 does not exist
// System.out.println(drinks[-1]);   // negative index is never valid

// Common mistake: off-by-one using length instead of length-1
int lastIndex = drinks.length - 1;   // correct: 2
System.out.println("Last index:  " + lastIndex);
System.out.println("Last drink:  " + drinks[lastIndex]);
````

## Vocabulary Recap

| Term | Definition |
|---|---|
| **Array** | Fixed-size, ordered collection of values of one type |
| **Index** | Zero-based integer position of an element (0 to length-1) |
| **`length`** | Array field returning the number of elements; NOT a method (no parentheses) |
| **Initializer list** | Curly-brace syntax to declare and fill an array at once |
| **Default value** | Value assigned to uninitialized slots: 0, 0.0, false, or null |
| **`ArrayIndexOutOfBoundsException`** | Runtime error from using an index < 0 or >= length |
