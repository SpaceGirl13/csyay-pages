## 2D Array Traversals

> One loop wasn't enough. Now we nest them.

---


## Core Concept

To visit every cell in a 2D array you need **two nested loops** --
the outer loop moves through rows, the inner loop moves through columns.

```java
for (int row = 0; row < arr.length; row++) {
    for (int col = 0; col < arr[row].length; col++) {
        // arr[row][col] is the current cell
    }
}
```

You can also traverse **column by column** (swap which loop is outer/inner)
or use an **enhanced for loop** over each row, then each element.

### Key Vocabulary

| Term | Meaning |
|---|---|
| **Nested loop** | A loop inside another loop; the inner loop completes fully for each outer iteration |
| **Row-by-row traversal** | Outer loop = rows; inner loop = columns; visits left-to-right, top-to-bottom |
| **Column-by-column traversal** | Outer loop = columns; inner loop = rows; visits top-to-bottom, left-to-right |
| **Enhanced for on 2D array** | `for (int[] row : arr)` then `for (int val : row)` |


---
## Part 1 - Row-by-Row Traversal


````java
String[] days   = {"Mon","Tue","Wed","Thu","Fri"};
String[] drinks = {"Matcha","Boba","Oolong","Taro"};

int[][] sales = {
    {42, 31, 18, 25},
    {38, 27, 22, 19},
    {55, 40, 15, 30},
    {49, 35, 20, 28},
    {63, 52, 25, 41}
};

// Print the full sales board
System.out.printf("%-5s", "");
for (String d : drinks) System.out.printf("%-8s", d);
System.out.println();
System.out.println("-".repeat(37));

for (int row = 0; row < sales.length; row++) {
    System.out.printf("%-5s", days[row]);
    for (int col = 0; col < sales[row].length; col++) {
        System.out.printf("%-8d", sales[row][col]);
    }
    System.out.println();
}
````

---
## Part 2 - Column-by-Column Traversal


````java
int[][] sales = {
    {42, 31, 18, 25},
    {38, 27, 22, 19},
    {55, 40, 15, 30},
    {49, 35, 20, 28},
    {63, 52, 25, 41}
};
String[] drinks = {"Matcha","Boba","Oolong","Taro"};

// Total sold per drink (column sums)
for (int col = 0; col < sales[0].length; col++) {
    int colTotal = 0;
    for (int row = 0; row < sales.length; row++) {
        colTotal += sales[row][col];    // same col, all rows
    }
    System.out.println(drinks[col] + " total this week: " + colTotal);
}
````

---
## Part 3 - Enhanced For Loop on 2D Arrays


````java
int[][] sales = {
    {42, 31, 18, 25},
    {38, 27, 22, 19},
    {55, 40, 15, 30},
    {49, 35, 20, 28},
    {63, 52, 25, 41}
};

// Enhanced for: outer = each row (int[] type), inner = each value in that row
int grandTotal = 0;
for (int[] row : sales) {
    for (int val : row) {
        grandTotal += val;
    }
}
System.out.println("Grand total drinks sold this week: " + grandTotal);

// Note: enhanced for cannot give you row/col indices.
// If you need indices, use the standard nested for loop.
````

## Vocabulary Recap

| Term | Definition |
|---|---|
| **Nested loop** | Loop inside a loop; inner completes fully for each outer iteration |
| **Row-by-row** | Outer iterates rows; inner iterates columns |
| **Column-by-column** | Outer iterates columns; inner iterates rows |
| **`for (int[] row : arr)`** | Enhanced for outer loop; each `row` is a 1D int array |
