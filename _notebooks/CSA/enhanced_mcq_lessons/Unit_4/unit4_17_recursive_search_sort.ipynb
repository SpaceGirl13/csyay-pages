## Recursive Searching and Sorting

> Binary search and merge sort: the recursive versions.

---


## Core Concept

The searching and sorting algorithms from 4.14-4.15 can be rewritten recursively.
The logic is the same -- the structure is just expressed through method calls instead of loops.

### Recursive Binary Search
- **Base cases:** target found at mid, OR low > high (not found)
- **Recursive case:** call on the left half OR the right half

### Merge Sort (recursive sort)
- **Base case:** array of size 0 or 1 is already sorted
- **Recursive case:** split in half, sort each half recursively, then merge them together
- Runtime: O(n log n) -- much faster than selection/insertion sort on large data

| Algorithm | Runtime | Strategy |
|---|---|---|
| Recursive binary search | O(log n) | Halve the search range via recursive call |
| Merge sort | O(n log n) | Split, sort halves recursively, merge |
| Selection / insertion sort | O(n^2) | Loops only |

### Key Vocabulary

| Term | Meaning |
|---|---|
| **Recursive binary search** | Binary search expressed as a method that calls itself on one half |
| **Merge sort** | Divide-and-conquer sort: split, recursive sort, merge |
| **Merge** | Combine two sorted arrays into one sorted array |
| **Divide and conquer** | Break a problem in half, solve each half, combine results |
| **O(n log n)** | Merge sort runtime; faster than O(n^2) for large inputs |

## Part 1 - Recursive Binary Search


```java
// Same logic as the iterative version -- now using recursion instead of a while loop
String[] menu = {"Classic Boba","Jasmine Tea","Matcha Latte","Oolong Tea","Taro Boba"};

public static int binarySearch(String[] arr, String target, int low, int high) {
    // BASE CASE 1: search range is empty -- not found
    if (low > high) return -1;

    int mid = (low + high) / 2;
    int cmp = target.compareTo(arr[mid]);

    // BASE CASE 2: found it
    if (cmp == 0) return mid;

    // RECURSIVE CASE: search left or right half
    if (cmp < 0) {
        System.out.println("  '" + target + "' < '" + arr[mid] + "' -- search LEFT");
        return binarySearch(arr, target, low, mid - 1);     // left half
    } else {
        System.out.println("  '" + target + "' > '" + arr[mid] + "' -- search RIGHT");
        return binarySearch(arr, target, mid + 1, high);    // right half
    }
}

System.out.println("Searching for 'Oolong Tea':");
int idx = binarySearch(menu, "Oolong Tea", 0, menu.length - 1);
System.out.println("Found at index: " + idx);

System.out.println("
Searching for 'Espresso':");
System.out.println("Result: " + binarySearch(menu, "Espresso", 0, menu.length - 1));
```

## Part 2 - Merge Sort


```java
// Merge sort on drink prices
// Step 1: the merge helper -- combines two sorted halves into one sorted array
public static double[] merge(double[] left, double[] right) {
    double[] result = new double[left.length + right.length];
    int i = 0, j = 0, k = 0;
    while (i < left.length && j < right.length) {
        if (left[i] <= right[j]) result[k++] = left[i++];
        else                     result[k++] = right[j++];
    }
    while (i < left.length)  result[k++] = left[i++];
    while (j < right.length) result[k++] = right[j++];
    return result;
}

// Step 2: the recursive sort
public static double[] mergeSort(double[] arr) {
    if (arr.length <= 1) return arr;               // BASE CASE: already sorted

    int mid = arr.length / 2;

    // Split
    double[] left  = java.util.Arrays.copyOfRange(arr, 0, mid);
    double[] right = java.util.Arrays.copyOfRange(arr, mid, arr.length);

    // Recursive sort each half
    left  = mergeSort(left);
    right = mergeSort(right);

    // Merge sorted halves
    return merge(left, right);
}

double[] prices = {6.50, 4.50, 5.75, 6.00, 5.50, 4.00};
System.out.print("Before: ");
for (double p : prices) System.out.printf("$%.2f ", p);

double[] sorted = mergeSort(prices);
System.out.print("
After:  ");
for (double p : sorted) System.out.printf("$%.2f ", p);
System.out.println();
```

## Part 3 - Tracing Merge Sort


```java
// Visualize the split-and-merge structure
public static double[] mergeSortVerbose(double[] arr, int depth) {
    String indent = "  ".repeat(depth);
    System.out.print(indent + "mergeSort(");
    for (double v : arr) System.out.printf("%.0f ", v);
    System.out.println(")");

    if (arr.length <= 1) {
        System.out.println(indent + "-> base case, return as-is");
        return arr;
    }

    int mid    = arr.length / 2;
    double[] L = mergeSortVerbose(java.util.Arrays.copyOfRange(arr, 0, mid),   depth + 1);
    double[] R = mergeSortVerbose(java.util.Arrays.copyOfRange(arr, mid, arr.length), depth + 1);
    double[] merged = merge(L, R);

    System.out.print(indent + "merged -> ");
    for (double v : merged) System.out.printf("%.0f ", v);
    System.out.println();
    return merged;
}

double[] prices = {4.50, 6.50, 5.75, 6.00};
System.out.println("=== Merge Sort Trace ===");
mergeSortVerbose(prices, 0);
```
