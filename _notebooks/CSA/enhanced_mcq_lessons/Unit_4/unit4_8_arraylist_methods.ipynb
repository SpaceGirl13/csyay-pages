## Core Concept

An `ArrayList` is a **resizable** list of objects.
Unlike arrays, it grows and shrinks dynamically -- no fixed size needed at creation.

### ArrayList vs. Array

| Feature | Array | ArrayList |
|---|---|---|
| Size | Fixed at creation | Grows/shrinks automatically |
| Syntax | `int[]` | `ArrayList<Integer>` |
| Access | `arr[i]` | `list.get(i)` |
| Length | `arr.length` | `list.size()` |
| Stores primitives? | Yes | No -- must use wrapper types |
| Import needed? | No | `import java.util.ArrayList;` |

### Key Methods

| Method | What it does |
|---|---|
| `add(E obj)` | Appends obj to the end |
| `add(int i, E obj)` | Inserts obj at index i; shifts elements right |
| `get(int i)` | Returns element at index i |
| `set(int i, E obj)` | Replaces element at index i; returns old value |
| `remove(int i)` | Removes element at index i; shifts remaining elements left |
| `size()` | Returns the number of elements |
| `contains(Object o)` | Returns true if o is in the list |
| `indexOf(Object o)` | Returns index of first occurrence; -1 if not found |
| `isEmpty()` | Returns true if size() == 0 |
| `clear()` | Removes all elements |

### Key Vocabulary

| Term | Meaning |
|---|---|
| **`ArrayList<E>`** | A generic resizable list; E is the element type (must be a class, not a primitive) |
| **Generic type parameter** | The `<E>` in `ArrayList<E>`; specifies what type the list holds |
| **`add`** | Appends or inserts an element |
| **`remove`** | Removes by index (shifts remaining elements) |
| **`size()`** | Returns current number of elements (a method, not a field) |
| **Index shifting** | When an element is inserted or removed, all elements to its right shift |

## Part 1 - Creating an ArrayList and Basic Methods


````java
import java.util.ArrayList;

// Create an ArrayList of Strings (drink names)
ArrayList<String> menu = new ArrayList<String>();

// add(E obj) -- append to end
menu.add("Matcha Latte");
menu.add("Classic Boba");
menu.add("Oolong Tea");
menu.add("Taro Boba");

System.out.println("Menu: "  + menu);
System.out.println("Size: "  + menu.size());
System.out.println("First: " + menu.get(0));
System.out.println("Last:  " + menu.get(menu.size() - 1));
System.out.println("Has Oolong? " + menu.contains("Oolong Tea"));
````

## Part 2 - add, set, and remove


````java
import java.util.ArrayList;

ArrayList<String> menu = new ArrayList<String>();
menu.add("Matcha Latte");
menu.add("Classic Boba");
menu.add("Oolong Tea");
menu.add("Taro Boba");
System.out.println("Original:  " + menu);

// add(int i, E obj) -- insert at index; elements to the right shift right
menu.add(1, "Strawberry Matcha");
System.out.println("After insert at 1: " + menu);

// set(int i, E obj) -- replace element at index; returns the old value
String old = menu.set(3, "Lychee Boba");
System.out.println("Replaced: " + old + "  -->  Lychee Boba");
System.out.println("After set: " + menu);

// remove(int i) -- remove by index; elements to the right shift left
String removed = menu.remove(0);
System.out.println("Removed: " + removed);
System.out.println("After remove: " + menu);
````

## Part 3 - ArrayList of Numbers (Wrapper Types)


````java
import java.util.ArrayList;

// ArrayList cannot store primitives -- use Integer, Double, Boolean
ArrayList<Double> prices = new ArrayList<Double>();

// Autoboxing: doubles are automatically wrapped as Double objects
prices.add(5.75);
prices.add(6.00);
prices.add(4.50);
prices.add(6.50);

System.out.println("Prices: " + prices);

// Compute total using unboxing (Double -> double automatically)
double total = 0.0;
for (double p : prices) {       // enhanced for with unboxing
    total += p;
}
System.out.printf("Total:   $%.2f%n", total);
System.out.printf("Average: $%.2f%n", total / prices.size());

// Remove by value: remove(Object) -- must pass an Object, not an int index!
prices.remove(Double.valueOf(6.00));    // removes the first occurrence of 6.00
System.out.println("After removing $6.00: " + prices);
````

## Part 4 - Index Shifting (Major MCQ Trap)


````java
import java.util.ArrayList;

// When you remove element at index i, everything to the RIGHT shifts LEFT by one.
// If you loop forward and remove, you skip the element that shifted into the removed slot.

ArrayList<String> drinks = new ArrayList<String>();
drinks.add("Matcha");    // index 0
drinks.add("Boba");      // index 1
drinks.add("Oolong");    // index 2
drinks.add("Taro");      // index 3

System.out.println("Before: " + drinks);

// Safe removal going BACKWARDS avoids the shifting problem
for (int i = drinks.size() - 1; i >= 0; i--) {
    if (drinks.get(i).contains("o")) {    // remove drinks whose name contains 'o'
        drinks.remove(i);
    }
}

System.out.println("After removing drinks with 'o': " + drinks);
````

````java
import java.util.ArrayList;

// Tracing index shifts -- a very common MCQ pattern
ArrayList<String> list = new ArrayList<String>();
list.add("A");   // 0
list.add("B");   // 1
list.add("C");   // 2
list.add("D");   // 3

System.out.println("Before: " + list);

list.remove(1);  // removes "B"; "C" shifts to index 1, "D" to index 2
System.out.println("After remove(1): " + list);
System.out.println("  get(1) is now: " + list.get(1));   // "C", not "B"

list.add(0, "Z");   // inserts "Z" at front; "A" shifts to 1, "C" to 2, "D" to 3
System.out.println("After add(0,Z): " + list);
````

## Vocabulary Recap

| Term | Definition |
|---|---|
| **`ArrayList<E>`** | Resizable list storing objects of type E; requires `import java.util.ArrayList` |
| **Generic type** | `<E>` specifies the element type; must be a class (not a primitive) |
| **`add(E)`** | Appends to the end |
| **`add(int, E)`** | Inserts at the given index; elements to the right shift right |
| **`remove(int)`** | Removes element at the given index; elements shift left |
| **`set(int, E)`** | Replaces element at the given index; returns the old element |
| **`size()`** | Returns current element count (method, not field -- unlike array `.length`) |
| **Index shifting** | After insert or remove, indices of neighboring elements change |
