## Core Concept

Programs often read data from external **text files** instead of hardcoding values.
Java reads files using the `Scanner` class with a `File` object.

### Basic pattern
```java
import java.util.Scanner;
import java.io.File;
import java.io.FileNotFoundException;

Scanner fileScanner = new Scanner(new File("menu.txt"));
while (fileScanner.hasNextLine()) {
    String line = fileScanner.nextLine();
    // process line
}
fileScanner.close();
```

### Reading methods on Scanner

| Method | Returns | Use for |
|---|---|---|
| `hasNextLine()` | boolean | Check if another line exists |
| `nextLine()` | String | Read the entire next line |
| `hasNextInt()` | boolean | Check if next token is an int |
| `nextInt()` | int | Read next int token |
| `hasNextDouble()` | boolean | Check if next token is a double |
| `nextDouble()` | double | Read next double token |

### Key Vocabulary

| Term | Meaning |
|---|---|
| **Text file** | A plain file storing data as readable characters |
| **`File`** | Java class representing a file path on disk |
| **`Scanner`** | Class for reading input -- from keyboard, String, or File |
| **`FileNotFoundException`** | Checked exception thrown when the specified file does not exist |
| **`throws` clause** | Declares that a method may throw a checked exception |
| **Token** | A chunk of text separated by whitespace; `next()` reads one token at a time |
| **Delimiter** | Character(s) that separate tokens (default: whitespace) |

## Part 1 - Creating and Reading a Text File


````java
import java.io.File;
import java.io.FileWriter;
import java.io.PrintWriter;
import java.util.Scanner;

// Step 1: Write a sample menu file so we have something to read
PrintWriter writer = new PrintWriter(new FileWriter("cafe_menu.txt"));
writer.println("Matcha Latte,5.75,true");
writer.println("Classic Boba,6.00,false");
writer.println("Oolong Tea,4.50,true");
writer.println("Taro Boba,6.50,false");
writer.println("Strawberry Matcha,5.50,true");
writer.close();

System.out.println("cafe_menu.txt created.");
````

````java
import java.io.File;
import java.util.Scanner;

// Step 2: Read the file line by line
Scanner fileScanner = new Scanner(new File("cafe_menu.txt"));

System.out.println("--- Cafe Menu (raw file) ---");
while (fileScanner.hasNextLine()) {
    String line = fileScanner.nextLine();
    System.out.println(line);
}
fileScanner.close();
````

## Part 2 - Parsing CSV Data from a File


````java
import java.io.File;
import java.util.Scanner;

// Each line is "name,price,isVegan" -- split on the comma to parse fields
Scanner fileScanner = new Scanner(new File("cafe_menu.txt"));

System.out.printf("%-20s %6s  %s%n", "Drink", "Price", "Vegan");
System.out.println("-------------------------------------");

while (fileScanner.hasNextLine()) {
    String   line   = fileScanner.nextLine();
    String[] parts  = line.split(",");         // split on comma

    String  name    = parts[0];
    double  price   = Double.parseDouble(parts[1]);   // String -> double
    boolean isVegan = Boolean.parseBoolean(parts[2]); // String -> boolean

    System.out.printf("%-20s $%.2f   %s%n", name, price, isVegan ? "Yes" : "No");
}
fileScanner.close();
````

## Part 3 - Loading File Data into an Array


````java
import java.io.File;
import java.util.Scanner;

// First pass: count lines so we know the array size
Scanner counter = new Scanner(new File("cafe_menu.txt"));
int lineCount = 0;
while (counter.hasNextLine()) { counter.nextLine(); lineCount++; }
counter.close();

// Second pass: fill arrays
String[] names    = new String[lineCount];
double[] prices   = new double[lineCount];
boolean[] isVegan = new boolean[lineCount];

Scanner reader = new Scanner(new File("cafe_menu.txt"));
int i = 0;
while (reader.hasNextLine()) {
    String[] parts = reader.nextLine().split(",");
    names[i]    = parts[0];
    prices[i]   = Double.parseDouble(parts[1]);
    isVegan[i]  = Boolean.parseBoolean(parts[2]);
    i++;
}
reader.close();

// Now use the arrays
System.out.println("Items loaded: " + names.length);
System.out.println("First item:   " + names[0] + " at $" + prices[0]);
System.out.println("Last item:    " + names[names.length - 1]);

// Find cheapest
double min = prices[0];
String cheapest = names[0];
for (int j = 1; j < prices.length; j++) {
    if (prices[j] < min) { min = prices[j]; cheapest = names[j]; }
}
System.out.println("Cheapest:     " + cheapest + " ($" + min + ")");
````

## Part 4 - FileNotFoundException


````java
import java.io.File;
import java.util.Scanner;

// Attempting to open a file that does not exist throws FileNotFoundException
// The method must declare 'throws FileNotFoundException' or handle it with try-catch
try {
    Scanner bad = new Scanner(new File("does_not_exist.txt"));
    bad.close();
} catch (java.io.FileNotFoundException e) {
    System.out.println("Caught FileNotFoundException: " + e.getMessage());
    System.out.println("Program continues running after the catch block.");
}
````

## Vocabulary Recap

| Term | Definition |
|---|---|
| **Text file** | A plain file where data is stored as readable characters |
| **`File`** | Java class that represents a file path; passed to `Scanner` to read from disk |
| **`Scanner`** | Reads input from a file, String, or keyboard token by token or line by line |
| **`FileNotFoundException`** | Checked exception; must be declared with `throws` or caught with try-catch |
| **Token** | A chunk of text delimited by whitespace; read with `next()`, `nextInt()`, etc. |
| **`split(",")`** | String method that divides a line on a delimiter and returns a String array |
