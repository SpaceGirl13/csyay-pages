## Core Concept

A **data set** is a structured collection of related values.
In Java, we use **arrays** (and later `ArrayList`) to store multiple values of the same type.

Before writing array code, it helps to understand:
- What the data represents
- How it is organized (rows? columns? labeled?)
- What questions we want to answer from it

### Cafe Data Set Example

Imagine the cafe tracks daily drink sales for one week:

| Day | Matcha | Boba | Oolong | Taro |
|---|---|---|---|---|
| Mon | 42 | 31 | 18 | 25 |
| Tue | 38 | 27 | 22 | 19 |
| Wed | 55 | 40 | 15 | 30 |
| Thu | 49 | 35 | 20 | 28 |
| Fri | 63 | 52 | 25 | 41 |

We can answer questions like: Which day had the most sales? Which drink is most popular?

### Key Vocabulary

| Term | Meaning |
|---|---|
| **Data set** | A structured collection of related values |
| **Array** | A fixed-size, ordered collection of values of the same type |
| **Element** | A single value stored in an array |
| **Index** | The position of an element in an array; starts at 0 |
| **Parallel arrays** | Two or more arrays where the element at index i in each array belongs to the same record |
| **Data type** | The kind of value stored (int, double, String, boolean, etc.) |

## Part 1 - Representing a Data Set in Java


```java
// Representing one row of cafe data as parallel arrays
// Index 0 = Monday, 1 = Tuesday, ... 4 = Friday

String[] days      = {"Monday", "Tuesday", "Wednesday", "Thursday", "Friday"};
int[]    matchaSales = {42, 38, 55, 49, 63};
int[]    bobaSales   = {31, 27, 40, 35, 52};
int[]    oolongSales = {18, 22, 15, 20, 25};
int[]    taroSales   = {25, 19, 30, 28, 41};

// These are PARALLEL arrays: index 2 in each array all refer to Wednesday
System.out.println("Wednesday matcha sales: " + matchaSales[2]);
System.out.println("Wednesday boba sales:   " + bobaSales[2]);
System.out.println("Wednesday oolong sales: " + oolongSales[2]);
System.out.println("Wednesday taro sales:   " + taroSales[2]);
```

```java
// Computing a meaningful statistic from the data set
// Total sales for each day across all drinks

for (int i = 0; i < days.length; i++) {
    int dailyTotal = matchaSales[i] + bobaSales[i] + oolongSales[i] + taroSales[i];
    System.out.println(days[i] + ": " + dailyTotal + " total drinks sold");
}
```

## Part 2 - Choosing the Right Data Type

Picking the wrong type wastes memory or loses precision.

| What you are storing | Use |
|---|---|
| Whole number count (drinks sold) | `int` |
| Price, percentage, average | `double` |
| Name, label, description | `String` |
| Yes/no flag | `boolean` |
| A collection of the above | Array of that type |


```java
// Cafe pricing data set -- prices need double, counts need int
String[] menuItems = {"Matcha Latte", "Classic Boba", "Oolong Tea", "Taro Boba"};
double[] prices    = {5.75, 6.00, 4.50, 6.50};
int[]    unitsSold = {247, 185, 100, 163};
boolean[] isVegan  = {true, false, true, false};

// Print a simple sales report
System.out.println("Item               Price   Sold   Vegan");
System.out.println("------------------------------------------");
for (int i = 0; i < menuItems.length; i++) {
    System.out.printf("%-18s $%.2f   %-6d %s%n",
        menuItems[i], prices[i], unitsSold[i], isVegan[i] ? "Yes" : "No");
}
```

## Part 3 - Asking Questions of a Data Set

Most data set work involves answering one of these questions:

| Goal | Approach |
|---|---|
| Find the total | Add up all elements |
| Find the average | Total divided by count |
| Find max or min | Track the best seen so far while traversing |
| Find an item | Check each element until a match is found |
| Filter | Collect elements that meet a condition |


```java
// What is the average price on the menu?
String[] menuItems = {"Matcha Latte", "Classic Boba", "Oolong Tea", "Taro Boba"};
double[] prices    = {5.75, 6.00, 4.50, 6.50};

double total = 0;
for (int i = 0; i < prices.length; i++) {
    total += prices[i];
}
double average = total / prices.length;
System.out.printf("Average menu price: $%.2f%n", average);

// Which item is cheapest?
double minPrice = prices[0];
String cheapest = menuItems[0];
for (int i = 1; i < prices.length; i++) {
    if (prices[i] < minPrice) {
        minPrice = prices[i];
        cheapest = menuItems[i];
    }
}
System.out.println("Cheapest item: " + cheapest + " at $" + minPrice);
```
