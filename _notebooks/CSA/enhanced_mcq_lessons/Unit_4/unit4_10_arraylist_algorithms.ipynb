## Core Concept

The same algorithm patterns from arrays (search, max/min, filter, count) apply to ArrayLists.
The difference is syntax: `list.get(i)` instead of `arr[i]`, and `list.size()` instead of `arr.length`.

ArrayList also unlocks one new power arrays don't have: **building a filtered sublist** on the fly,
because you can add to an ArrayList without knowing its size upfront.

### Key patterns

| Goal | Approach |
|---|---|
| Find an item | Linear search; return index or -1 |
| Find max/min | Initialize to `list.get(0)`, compare the rest |
| Filter | Create a new ArrayList; add elements that pass the test |
| Count | Accumulate a counter with a conditional inside the loop |
| Remove matches | Traverse backwards and remove |

## Part 1 - Search and Find Max/Min


````java
import java.util.ArrayList;

ArrayList<String> menu   = new ArrayList<String>();
ArrayList<Double> prices = new ArrayList<Double>();
menu.add("Matcha Latte");      prices.add(5.75);
menu.add("Classic Boba");      prices.add(6.00);
menu.add("Oolong Tea");        prices.add(4.50);
menu.add("Taro Boba");         prices.add(6.50);
menu.add("Strawberry Matcha"); prices.add(5.50);

// Search: find index of a drink by name
public static int findDrink(ArrayList<String> menu, String target) {
    for (int i = 0; i < menu.size(); i++) {
        if (menu.get(i).equals(target)) return i;
    }
    return -1;
}

int idx = findDrink(menu, "Oolong Tea");
System.out.println("Oolong Tea at index: " + idx);
System.out.println("Price: $" + prices.get(idx));
System.out.println("Not found: " + findDrink(menu, "Espresso"));

// Max price
double maxPrice = prices.get(0);
String maxName  = menu.get(0);
for (int i = 1; i < prices.size(); i++) {
    if (prices.get(i) > maxPrice) {
        maxPrice = prices.get(i);
        maxName  = menu.get(i);
    }
}
System.out.printf("Most expensive: %s ($%.2f)%n", maxName, maxPrice);
````

## Part 2 - Filtering into a New ArrayList


````java
import java.util.ArrayList;

// The manager wants to see only drinks under $6.00 for a budget menu
ArrayList<String> menu   = new ArrayList<String>();
ArrayList<Double> prices = new ArrayList<Double>();
menu.add("Matcha Latte");      prices.add(5.75);
menu.add("Classic Boba");      prices.add(6.00);
menu.add("Oolong Tea");        prices.add(4.50);
menu.add("Taro Boba");         prices.add(6.50);
menu.add("Strawberry Matcha"); prices.add(5.50);

ArrayList<String> budgetMenu = new ArrayList<String>();

for (int i = 0; i < menu.size(); i++) {
    if (prices.get(i) < 6.00) {
        budgetMenu.add(menu.get(i) + " ($" + prices.get(i) + ")");
    }
}

System.out.println("Budget Menu (under $6):");
for (String item : budgetMenu) {
    System.out.println("  " + item);
}
````

## Part 3 - Remove Duplicates


````java
import java.util.ArrayList;

// A customer accidentally submitted the same order three times
// Remove all duplicates, keep only the first occurrence
ArrayList<String> orders = new ArrayList<String>();
orders.add("Matcha Latte");
orders.add("Taro Boba");
orders.add("Matcha Latte");   // duplicate!
orders.add("Oolong Tea");
orders.add("Taro Boba");      // duplicate!
orders.add("Classic Boba");

System.out.println("Before dedup: " + orders);

ArrayList<String> seen = new ArrayList<String>();
for (int i = orders.size() - 1; i >= 0; i--) {
    if (seen.contains(orders.get(i))) {
        orders.remove(i);        // already seen -- remove this later duplicate
    } else {
        seen.add(orders.get(i)); // first time seeing it -- keep it
    }
}

System.out.println("After dedup:  " + orders);
````

## Vocabulary Recap

| Term | Definition |
|---|---|
| **Filter** | Build a new collection containing only elements that satisfy a condition |
| **`contains(obj)`** | Returns true if the ArrayList holds that object (uses `.equals()`) |
| **Deduplication** | Removing duplicate values, keeping only the first occurrence |
| **Parallel ArrayLists** | Two lists where index i in each list refers to the same record |
