# 3.4 — Else-If Statements 🧋

### AP CSA · Unit 3: Boolean Expressions and if Statements
**Boba Cafe Series · Lesson 3.4**

---

**Else If Statements**: Used when you have multiple conditions that need to be checked sequentially (like choosing drink sizes or customizing sweetness levels).

**Flow of Execution**: Each condition is evaluated in the order written. The first true condition's code runs, and the rest are skipped.

**Structure**:

- Start with a single `if` statement.
- Follow with as many `else if` statements as needed.
- Optionally end with one `else` to handle any remaining cases.

**Key Concept**: The order of conditions matters. More specific conditions should come before broader ones to ensure accurate results.

```java
// Cafe Logic Example: Loyalty Stars Tiers
int loyaltyStars = 17; 
System.out.println("Current loyalty stars: " + loyaltyStars);

if (loyaltyStars >= 20) 
{
    System.out.println("VIP Status: Free premium boba drink!");
}
else if (loyaltyStars >= 15) {
    System.out.println("Gold Status: 50% off your next order.");
} 
else if (loyaltyStars >= 10) {
    System.out.println("Silver Status: Free boba topping.");
}
else if (loyaltyStars >= 5) {
    System.out.println("Bronze Status: Free upsize.");
}
```
