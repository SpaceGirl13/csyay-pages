## Purpose of Else Statements in the Cafe
Else statements: Handles what happens when the `if` condition is false.
Structure of If-Else:
- `if` statement with a condition (e.g., Does the customer want ice?).
- `else` statement without a condition (e.g., What to do if they don't want ice).
- Both parts have code blocks surrounded by `{}`.

- *Note: if there is only one instruction in the if statement technically brackets aren't mandatory, but it's good practice to use them because generally there are more instructions in the statements.*


```java
// example of a simple if-else statement: 
// things to keep in mind: 
// Always put a semicolon after a statement 
// make sure you use double quotes for your message 

int orderNumber = 2; // change this number to see what happens when it's odd

if (orderNumber % 2 == 0) {
    System.out.println("Order number is even - prepare at Station A");
} else {
    System.out.println("Order number is odd - prepare at Station B");
}
```

```java
// slightly more complex statement because there are 2 messages 

int loyaltyStars = 17; // adjust the stars to see the difference in what it prints out

System.out.println("Current loyalty stars: " + loyaltyStars);

if (loyaltyStars >= 10) {
    System.out.println("You have enough stars for a free boba!");
    System.out.println("Enjoy your drink!");
} else {
    System.out.println("You don't have enough stars for a free drink yet.");
    System.out.println("Keep sipping!");
}
```
