# while Loops in the Boba Cafe 🧋

---

- A `while` loop is a fundamental control structure in programming used for repeated execution of a block of code as long as a condition is true [cite: 15].
- The loop starts by evaluating the condition (e.g., checking if there are still boba pearls left in the pot). If true, the code inside runs [cite: 15].
- After each iteration, the condition is re-evaluated; if still true, the loop continues [cite: 15]. If false initially, it never executes [cite: 15].
- While loops are used when you don't know in advance how many times the loop needs to execute (like waiting for a customer's order queue to clear) [cite: 15].
- Watch out for infinite loops if the condition never becomes false (e.g., forgetting to scoop out pearls) [cite: 15].
- It's essential to update the loop control variable within the loop [cite: 15].
- You can always break out of a while loop prematurely using the `break` statement when closing the shop [cite: 15].

## Example of While Loops: Boba Cup Display

```java
public class BobaCupPyramid {
    public static void main(String[] args) {
        int height = 5; // number of rows in our boba display
        int row = 1;

        while (row <= height) {
            int spaces = height - row;
            int pearls = 2 * row - 1;

            // Print leading spaces
            int spaceCount = spaces;
            while (spaceCount > 0) {
                System.out.print(" ");
                spaceCount--;
            }

            // Print pearls/stars for the boba display
            int pearlCount = pearls;
            while (pearlCount > 0) {
                System.out.print("🧋");
                pearlCount--;
            }

            System.out.println(); // Move to the next row
            row++;
        }
    }
}
BobaCupPyramid.main(null);
```

# 4.4 Nested Iteration in the Cafe ☕

**nested iteration** [cite: 15]
<details>occurs when we have a loop inside of another loop, similar to nested conditional statements in unit 3 [cite: 15].

When you have one loop inside another, the inner loop finishes all its rounds for every single round of the outer loop [cite: 15]. If the outer loop runs $n$ times and the inner loop runs $m$ times per round, the total iterations equal $m \times n$ [cite: 15]. 

*Boba Cafe Context*: Think of the outer loop as brewing batches of tea, and the inner loop as filling individual cups for each batch!

```java
public class BobaBatchDemo {
    public static void main(String[] args) {
        int n = 3; // number of tea batches (outer loop)
        int m = 2; // number of cups filled per batch (inner loop)

        // Nested loops for boba production
        for (int i = 1; i <= n; i++) {
            System.out.println("Brewing Tea Batch # " + i);
            for (int j = 1; j <= m; j++) {
                System.out.println(" -> Filling Cup # " + j + " with boba pearls 🧋");
            }
        }
    }
}
BobaBatchDemo.main(null);
```

```java
// CODE EXAMPLE #2 (for loop)
public class BobaInventoryCheck {
    public static void main(String[] args) {
        int count = 0;
        for (int k = 4; k < 30; k+=3)
        {
            count++; // statement 3
        }
        System.out.println("Statement 3 executed: " + count + " times.");
    }
}
BobaInventoryCheck.main(null);
```

<b>How many times will statement 3 execute [cite: 15]?</b> [cite: 15]

Answer: Let's trace `k` values: 4, 7, 10, 13, 16, 19, 22, 25, 28 (9 times).

```java
// Rewrite the code segment below to have a faster run-time [cite: 15] based on statement execution counts [cite: 15]
public class OptimizedBobaTimer {
    public static void main(String[] args) {
        // Original code checks every single number up to 135
        for (int k = 0; k < 135; k += 5)
        {
            System.out.println("Restocking boba at interval: " + k);
        }
    }
}
OptimizedBobaTimer.main(null);
```

```java
// CODE EXAMPLE #3 (while loop for random customer orders)
public class RandomOrderCheck {
    public static void main(String[] args) {
        int orderTicket = (int)(Math.random() * 10);
        while (orderTicket % 2 != 0)
        {
            orderTicket = (int)(Math.random() * 10); // statement 4 [cite: 15]
        }
        System.out.println("Valid even ticket processed: " + orderTicket);
    }
}
RandomOrderCheck.main(null);
```

<b>What is the min/max number of times statement 4 [cite: 15] will execute [cite: 15]?</b> [cite: 15]

Answer: 
- **Min**: 0 times (if the very first random number is even).
- **Max**: Infinite theoretical maximum (since `Math.random()` can continuously generate odd numbers, though practically finite).

```java
// CODE EXAMPLE #4 (nested loop for boba station grid)
public class BobaStationGrid {
    public static void main(String[] args) {
        for (int outer = 0; outer < 3; outer++)
        {
            for (int inner = 0; inner < 4; inner++)
            {
                // statement #5 [cite: 15]
                System.out.println("Station row " + outer + ", col " + inner);
            }
        }
    }
}
BobaStationGrid.main(null);
```
