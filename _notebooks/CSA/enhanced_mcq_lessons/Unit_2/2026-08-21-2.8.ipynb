# 4.2 for Loops in the Boba Cafe 🧋

- An iterative statement that checks for a condition [cite: 17].
- Repeatedly executes a block of code as long as the condition is met [cite: 17] (e.g., sealing a fixed number of boba cups).
- The condition specifies the exact amount of times to loop [cite: 17].

### 🧋 for Loops vs. while Loops
- **while Loops**: Use when the number of iterations is **unknown** [cite: 17] (e.g., waiting for an unpredictable customer rush to clear).
- **for Loops**: Use when the number of iterations is **known** [cite: 17] (e.g., sealing exactly 5 orders of milk tea).

```java
// Example: while loop tracking boba cup preparation
public class BobaWhileDemo {
    public static void main(String[] args) {
        int i = 0;
        while (i < 5) {
            System.out.println("Preparing cup # " + i + " 🧋");
            i++;
        }
    }
}
BobaWhileDemo.main(null);
```

```java
// Example: for loop doing the exact same task more cleanly!
public class BobaForDemo {
    public static void main(String[] args) {
        for (int i = 0; i < 5; i++) {
            System.out.println("Sealing cup # " + i + " 🧋");
        }
    }
}
BobaForDemo.main(null);
```

- There are **three parts** in a `for` loop header: variable initialization, Boolean (conditional) expression, and increment/decrement statement [cite: 17].

### 🔍 Breakdown of the Header `for (int i = 0; i < 5; i++)`:
1. **Variable initialization** (`int i = 0`) [cite: 17]: Sets up the loop control variable before the loop starts.
2. **Boolean (conditional) expression** (`i < 5`) [cite: 17]: Defines the condition for the loop to run. The loop continues as long as `i` is less than 5 (running 5 total times: 0, 1, 2, 3, 4) [cite: 17].
3. **Increment/decrement statement** (`i++`) [cite: 17]: Increases or decreases the variable each time the code block executes [cite: 17].

*Tip*: The loop variable can also be used inside the code block for calculations, not just for counting iterations [cite: 17]!

```java
// Example: Calculating total boba pearls and even-batch pearl totals
public class BobaPearlTally {
    public static void main(String[] args) {
        int evenSum = 0;
        int totalSum = 0;

        for (int i = 0; i < 10; i++) {
            if (i % 2 == 0) {
                evenSum += i; // Tallying even batches
            }
            totalSum += i; // Tallying all batches
        }

        System.out.println("Even batch pearl sum: " + evenSum);
        System.out.println("Total pearl sum across all batches: " + totalSum);
    }
}
BobaPearlTally.main(null);
```
