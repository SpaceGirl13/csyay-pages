## 1. Boolean Basics (`true` and `false`)

In Java, a `boolean` data type can store only two values: `true` or `false`. Expressions that evaluate to these values are called **Boolean expressions**.

In our Boba Cafe, we use booleans to track operational states and customer criteria:
```java
boolean isStoreOpen = true;
boolean isSoldOut = false;
```

We also use **relational operators** to compare values and produce booleans:
- `>` (greater than)
- `<` (less than)
- `>=` (greater than or equal to)
- `<=` (less than or equal to)
- `==` (equal to)
- `!=` (not equal to)

## 2. Logical Operators: AND (`&&`), OR (`||`), and NOT (`!`)

When we need to check multiple conditions at once, we use logical operators to form **compound Boolean expressions**.

### Logical AND (`&&`)
- **What it does**: Combines two conditions.
- **Rule**: **Both** conditions must be `true` for the overall expression to be `true`.
- *Cafe Example*: Giving a free upgrade requires having enough loyalty stars **and** ordering a large size.

### Logical OR (`||`)
- **What it does**: Combines two conditions.
- **Rule**: **At least one** condition must be `true` for the overall expression to be `true`.
- *Cafe Example*: Applying a discount if it's your birthday **or** if you have a promo coupon.

### Logical NOT (`!`)
- **What it does**: Reversing a Boolean value.
- **Rule**: `true` becomes `false`, and `false` becomes `true`.
- *Cafe Example*: Checking if toppings are **not** sold out (`!isSoldOut`).

## 3. Order of Evaluation and Parentheses

When evaluating compound Boolean expressions, Java follows a strict order of operations:
1. Relational operators (`>`, `<`, `==`, etc.)
2. Logical NOT (`!`)
3. Logical AND (`&&`)
4. Logical OR (`||`)

**Important Rule**: `&&` is evaluated *before* `||`. 

To make complex conditions easier to read and to control the evaluation order explicitly, use **parentheses `()`**.

````java
// Example: Compound Boolean expressions in if statements for our Boba Cafe
public class BobaOrderProcessor {
    public static void main(String[] args) {
        int loyaltyStars = 12;
        boolean hasValidCoupon = true;
        boolean isStoreOpen = true;

        // Equivalent to checking if age >= 13 && hasPermission in basic CSA examples
        if (loyaltyStars >= 10 && hasValidCoupon) {
            System.out.println("Discount applied! Enjoy your boba reward.");
        } else {
            System.out.println("Standard pricing applies.");
        }

        // Using parentheses for clear evaluation order
        if (isStoreOpen && (loyaltyStars >= 15 || hasValidCoupon)) {
            System.out.println("Priority checkout lane unlocked!");
        }
    }
}

BobaOrderProcessor.main(null);
````

When breaking down complicated expressions into smaller parts, evaluate from left to right (respecting operator precedence and parentheses) to determine whether the final result is `true` or `false`.

# AP CSA Practice Questions

**1.** Consider the following code segment:
```java
boolean a = true;
boolean b = false;
boolean c = true;
boolean check = (a || b) && !c;
```
What is the value of `check` after execution?

&nbsp;&nbsp;(A) `true`
&nbsp;&nbsp;(B) `false`
&nbsp;&nbsp;(C) `null`
&nbsp;&nbsp;(D) Throws an exception

<details><summary>Answer</summary>

**(B) `false`**. 
*Explanation*: First, evaluate `(a || b)` which is `true || false`, resulting in `true`. Next, evaluate `!c`, which is `!true` resulting in `false`. Finally, `true && false` evaluates to `false`.
</details>
