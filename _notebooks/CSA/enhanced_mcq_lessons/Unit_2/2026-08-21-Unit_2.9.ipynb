Welcome to the Boba Cafe! Today, we are learning how to work with the `String` class. In Java, a String is just a sequence of characters—or in our case, the name of a delicious drink!

## 🎯 Learning Objectives
- Understand what a String is.
- Use common String methods (`length`, `substring`, `indexOf`, `equals`).
- Manipulate and compare text.

## 1. What Is a String?
A String is an object that represents a sequence of characters.

Think of it as a label on a boba cup.

```java
String drink = "Strawberry Milk Tea";
System.out.println(drink);
```

## 2. The `.length()` Method
The `.length()` method tells you how many characters (including spaces!) are in your String.

```java
String drink = "Milk Tea";
// Counts 8 characters (M-i-l-k- -T-e-a)
System.out.println("Drink name length: " + drink.length());
```

## 3. The `.substring()` Method
`substring(start, end)` allows you to get a slice of your String. 
⚠️ **Remember**: Start index is inclusive, End index is exclusive (stop *before* the end index).

```java
String drink = "Strawberry";
// Starts at 0 (S), stops before index 5 (y)
System.out.println(drink.substring(0, 5)); // Output: Strawb
```

## 4. The `.indexOf()` Method
`indexOf()` finds the position of the first occurrence of a character or substring.

```java
String drink = "Boba Milk Tea";
// 'B'(0) 'o'(1) 'b'(2) 'a'(3) ' '(4) 'M'(5)
System.out.println("Index of 'Milk': " + drink.indexOf("Milk"));
```

## 5. The `.equals()` Method
Always use `.equals()` to compare the *contents* of two Strings, never `==`!

```java
String drink1 = "Boba";
String drink2 = "Boba";
System.out.println("Do they match? " + drink1.equals(drink2));
```

## ⭐ AP CSA Cheat Sheet

| Method | Description |
| :--- | :--- |
| `.length()` | Number of characters |
| `.substring(s, e)` | Chars from `s` to `e-1` |
| `.indexOf(str)` | Index where `str` begins |
| `.equals(other)` | Compares contents |
| `.toUpperCase()` | Makes text UPPERCASE |
| `.toLowerCase()` | Makes text lowercase |
