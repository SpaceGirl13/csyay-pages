## Part 1 — Three building blocks, one recap

From Lesson 1.1:

| Structure | What it does |
|---|---|
| **Sequencing** | Steps happen in order, once each |
| **Selection** | A condition decides *which* steps happen |
| **Repetition** (iteration) | A step or group of steps happens *more than once* |

Every algorithm you will ever design — cafe workflow, Java program, or otherwise — is built entirely out of those three. This lesson zooms in on the two that let an algorithm stop being a straight line.

---

## Part 2 — Selection: describing a decision

> **Selection** is a point in an algorithm where the next step depends on a **condition** — the algorithm branches, and only one branch actually runs.

You don't need Java to describe selection. Plain English works fine, as long as it's precise (Lesson 1.1's rule for any algorithm):

> *"If the customer ordered oat milk, use the oat milk pitcher. Otherwise, use the dairy pitcher."*

That sentence is a complete, unambiguous algorithm. It has exactly one decision point and exactly two possible paths, and it's just as valid an algorithm as any Java program — language-independence, straight from Lesson 1.1.

### Drawing it as a flowchart

A **flowchart** represents selection with a diamond — one path in, and the path that's actually taken depends on the answer to a yes/no question.

```
                    ┌─────────────────────┐
                    │  Order comes in      │
                    └──────────┬───────────┘
                                │
                        ╱ Did they    ╲
                       ╱   ask for      ╲
                      ╱   oat milk?       ╲
                      ╲                   ╱
                        ╲                ╱
                     YES │            │ NO
                    ┌────▼───┐   ┌────▼─────┐
                    │  Use    │   │  Use      │
                    │  oat    │   │  dairy    │
                    │  milk   │   │  milk     │
                    └────┬───┘   └────┬─────┘
                          │            │
                          └─────┬──────┘
                                │
                        ┌───────▼────────┐
                        │  Continue order  │
                        └─────────────────┘
```

Both paths lead back to the **same place** afterward — the drink still gets made either way. That "paths reconnect" detail is worth noticing now; it's exactly how `if`/`else` behaves once you meet it formally in Lesson 2.3.

### A glimpse of where this is headed

**Preview only — Topic 2.3 teaches this syntax properly.** Just notice how closely it matches the English sentence and the flowchart above.

````java
public class SelectionPreview {
    public static void main(String[] args) {
        String milkType = "oat";

        if (milkType.equals("oat")) {
            System.out.println("Using the oat milk pitcher.");
        } else {
            System.out.println("Using the dairy milk pitcher.");
        }
    }
}

SelectionPreview.main(null);
````

Change `"oat"` to anything else and re-run — the decision point is the same, but a different path gets taken. That's selection, in eleven lines of Java instead of one sentence of English. Same algorithm, two representations.

## Part 3 — Repetition: describing a repeated action

> **Repetition** is a point in an algorithm where one or more steps happen **more than once**, continuing until some **stopping condition** is reached.

Plain English again:

> *"Seal one cup. If there are more cups in the tray, seal another. Keep going until every cup in the tray is sealed."*

Notice this sentence has two parts baked into it: **what repeats** (sealing one cup) and **when to stop** (every cup is sealed). Both parts are required — "keep sealing cups" alone isn't a real algorithm, because it never ends (Lesson 1.1's *finite* requirement).

### Drawing it as a flowchart

Repetition draws as a **loop-back arrow** — after the step runs, control returns to check the condition again, instead of moving straight ahead.

```
                    ┌──────────────────────┐
                    │  Start of shift        │
                    └───────────┬────────────┘
                                 │
                        ┌────────▼────────┐
                  ┌────►│  Seal one cup     │
                  │     └────────┬────────┘
                  │              │
                  │      ╱ More cups   ╲
                  │     ╱  left in the   ╲   YES
                  └────╲   tray?        ╱────┘
                         ╲              ╱
                           ╲          ╱
                          NO │
                        ┌────▼─────────┐
                        │  Tray done     │
                        └───────────────┘
```

That loop-back arrow is the entire idea of repetition in one picture: the algorithm keeps returning to the same step until the diamond finally answers "no."

### Two ways to know when to stop

Not every stopping condition looks the same. Two common shapes:

| Kind | Example | Stops when... |
|---|---|---|
| **Counted** | "Seal 6 cups." | A specific number of repetitions has happened |
| **Condition-based** | "Keep pouring until the cup reaches the fill line." | Some situation becomes true, and you don't know in advance exactly how many steps that takes |

Both are perfectly good algorithms. The tray-sealing example above is counted — you know there are exactly 6 cups. The pouring example is condition-based — you're watching the fill line, not counting ounces in your head.

### A glimpse of where this is headed

**Preview only — Topic 2.7+ teaches loop syntax properly.**

````java
public class RepetitionPreviewFor {
    public static void main(String[] args) {
        for (int cup = 1; cup <= 6; cup++) {
            System.out.println("Sealing cup " + cup);
        }
        System.out.println("Tray full.");
    }
}

RepetitionPreviewFor.main(null);
````

That's the *counted* stopping condition from the table above — Java is told exactly how many times to repeat. Here's the *condition-based* version, where Java doesn't know the count in advance and just keeps checking a condition instead.

````java
public class RepetitionPreviewWhile {
    public static void main(String[] args) {
        int ounces = 0;

        while (ounces < 12) {
            ounces = ounces + 4;
            System.out.println("Poured up to " + ounces + " oz");
        }

        System.out.println("Fill line reached.");
    }
}

RepetitionPreviewWhile.main(null);
````

Same idea as the flowchart's diamond — check a condition, repeat if it's still true, stop the moment it isn't. `for` and `while` are two different Java spellings for the exact same algorithmic pattern: repetition with a stopping condition.

## Part 4 — Selection and repetition together

Real workflows almost always need both. Here's the cafe's actual closing routine, in plain English:

> *"For each cup remaining on the counter: if it has a lid on already, move it straight to the tray. Otherwise, seal it first, then move it to the tray. Keep going until the counter is clear."*

Read that carefully — there's a **repetition** ("for each cup... keep going until the counter is clear") with a **selection** *nested inside it* ("if it has a lid already... otherwise..."). This is exactly the kind of layered structure you traced back in Lesson 1.1's `MorningRush` example, now named properly.

```
                    ┌─────────────────────┐
                    │  Any cups left on     │
                    │  the counter?          │
                    └──────────┬────────────┘
                          NO   │   YES
                    ┌──────────┘   └───────────┐
              ┌─────▼─────┐              ┌──────▼───────┐
              │  Done       │              │  Already has  │
              └────────────┘              │  a lid?        │
                                            └──────┬────────┘
                                             NO │      │ YES
                                          ┌──────▼┐  ┌──▼─────┐
                                          │ Seal it│  │ (skip)  │
                                          └───┬────┘  └───┬─────┘
                                              └─────┬──────┘
                                                     │
                                             ┌───────▼────────┐
                                             │  Move to tray    │
                                             └───────┬─────────┘
                                                      │
                                              (loop back to top)
```

### A glimpse of where this is headed

````java
public class CombinedPreview {
    public static void main(String[] args) {
        String[] counterCups = {"pearls", "jelly", "pearls"};

        for (int i = 0; i < counterCups.length; i++) {
            String topping = counterCups[i];

            if (topping.equals("pearls")) {
                System.out.println("Cup " + (i + 1) + ": scooping tapioca pearls");
            } else {
                System.out.println("Cup " + (i + 1) + ": scooping lychee jelly");
            }
        }

        System.out.println("Counter clear.");
    }
}

CombinedPreview.main(null);
````

> The `String[] counterCups` array is a Unit 4 tool, borrowed early just to make this preview realistic — you don't need to understand its syntax today.

Notice the shape: a repetition (the `for`) with a selection (the `if`/`else`) living inside it. That nesting relationship is the actual skill this lesson is building toward — being able to look at a real process and correctly identify *where* the loop is, *where* the decision is, and *how they fit together* — well before any of the Java syntax is formally taught.

## Part 5 — Three ways to say the same thing

This lesson has shown every algorithm in three different forms. That's deliberate, and it's worth stating outright:

| Representation | Best for |
|---|---|
| **Written language** | Quick to produce, easy to share, good for a first draft |
| **Flowchart / diagram** | Makes branching and looping visually obvious at a glance |
| **Code** (Java, eventually) | What a computer can actually run |

None of these is "more correct" than the others — from Lesson 1.1, an algorithm is language-independent. Professional software engineers routinely sketch a flowchart or write pseudocode **before** touching a keyboard, for exactly the reason this lesson exists: it's much easier to spot a missing stopping condition or a forgotten branch on paper than halfway through debugging real code.

# Self-Check: AP-style questions

**1.** Which best describes selection in an algorithm?

&nbsp;&nbsp;(A) Steps that repeat until a condition is met
&nbsp;&nbsp;(B) A point where the algorithm branches based on a condition, and only one branch executes
&nbsp;&nbsp;(C) Steps that always execute in the order they're written
&nbsp;&nbsp;(D) A step that never executes

<details><summary>Answer</summary>

**(B)**. Selection introduces a decision point — a condition determines which of several paths actually runs. (A) describes repetition, and (C) describes sequencing.
</details>

**2.** An algorithm for making toast says: "Put bread in the toaster. Keep toasting until it's golden brown." Which statement is true?

&nbsp;&nbsp;(A) This is a counted repetition, since toasting always takes a fixed number of steps
&nbsp;&nbsp;(B) This is a condition-based repetition, since the stopping point depends on the toast's appearance rather than a fixed count
&nbsp;&nbsp;(C) This algorithm contains no repetition at all
&nbsp;&nbsp;(D) This algorithm is invalid, since it never specifies a stopping condition

<details><summary>Answer</summary>

**(B)**. "Until it's golden brown" is a condition being checked, not a predetermined count — this is condition-based repetition. (D) is incorrect: a stopping condition *is* present, it's just based on a situation rather than a number.
</details>

**3.** In a flowchart, which shape typically represents a decision point?

&nbsp;&nbsp;(A) A rectangle &nbsp;&nbsp; (B) An oval &nbsp;&nbsp; (C) A diamond &nbsp;&nbsp; (D) An arrow

<details><summary>Answer</summary>

**(C)**. The diamond represents a yes/no question, with a different path leading out of each possible answer — the visual signature of selection.
</details>

**4.** Consider this algorithm: "For every customer in line, greet them, then take their order." Which structures are present?

&nbsp;&nbsp;(A) Sequencing only
&nbsp;&nbsp;(B) Selection only
&nbsp;&nbsp;(C) Repetition only
&nbsp;&nbsp;(D) Both repetition and sequencing, but no selection

<details><summary>Answer</summary>

**(D)**. "For every customer" is repetition. "Greet them, then take their order" is two steps happening in a fixed order — sequencing — repeated for each customer. No decision point is described, so there's no selection here.

**5.** Why is written language, a flowchart, and Java code all considered valid ways to represent the *same* algorithm?

&nbsp;&nbsp;(A) They aren't valid — only code counts as a real algorithm

# Closing time

### Vocabulary to know cold

| Term | One-line definition |
|---|---|
| Selection | A point where an algorithm branches based on a condition; only one path runs |
| Repetition | Steps that repeat until a stopping condition is reached |
| Stopping condition | The rule that determines when repetition ends |
| Counted repetition | Repetition where the number of repeats is known in advance |
| Condition-based repetition | Repetition that continues until a situation becomes true, count unknown in advance |
| Flowchart | A diagram representing an algorithm; diamonds for decisions, loop-back arrows for repetition |

### The three things that will show up on the exam

1. An algorithm can be described in **written language, a diagram, or code** — all equally valid.
2. Every repetition needs a genuine **stopping condition**, or it isn't a valid algorithm (Lesson 1.1's *finite* rule).
3. Selection and repetition **nest** inside each other constantly — learning to spot the nesting is the real skill here.
