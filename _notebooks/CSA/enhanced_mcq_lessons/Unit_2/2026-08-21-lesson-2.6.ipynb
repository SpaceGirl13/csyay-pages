# 2.6 — Comparing Objects 🧋

### AP CSA · Unit 2: Using Objects
**Boba Cafe Series · Lesson 2.6**

---

Unit 2.6 focuses on a critical concept in Java: **Comparing Objects**. It’s the difference between asking if two cups are the *same cup* versus asking if they contain the *same drink*.

## 🎯 Learning Objectives
- Understand how objects are compared in Java.
- Know the difference between `==` and `.equals()`.
- Understand reference equality.
- Use `.equals()` to compare object contents.
- Avoid common comparison mistakes.

## 1. Comparing Primitive Values
For primitive types (like `int`, `double`, `boolean`), `==` compares the **actual values**.

Think of this like counting the number of boba pearls in a cup. If both counts are 5, they are equal.

```java
int pearls1 = 5;
int pearls2 = 5;

System.out.println("Are the pearl counts equal? " + (pearls1 == pearls2));
// Output: true
```

## 2. Comparing Objects (The Memory Trap)
Objects work differently. When you use `new`, you are creating a *new reference* (a unique spot in memory).

Even if two drinks both say "Milk Tea", they are two **separate objects** (two different cups).

`==` checks if they are the **same object** (same memory address), not whether they contain the same text.

```java
String drink1 = new String("Milk Tea");
String drink2 = new String("Milk Tea");

// This will return false because they are different cups!
System.out.println("Are these the same object? " + (drink1 == drink2));
```

## 3. The `.equals()` Method
To compare the *contents* (ingredients) of objects, use `.equals()`.

`.equals()` asks: "Do these objects contain equivalent information?"

```java
String drink1 = new String("Milk Tea");
String drink2 = new String("Milk Tea");

// This returns true because the text inside is the same.
System.out.println("Do they have the same flavor? " + drink1.equals(drink2));
```

## 🌟 AP CSA Cheat Sheet

| Operator/Method | What it compares |
| :--- | :--- |
| `==` (with primitives) | Values |
| `==` (with objects) | References (Memory location) |
| `.equals()` | Object contents/equivalence |

### 🧋 Easy Memory Trick
- `==` → "Are these the **same thing** (same cup)?"
- `.equals()` → "Do these have the **same information** (same flavor)?"

**Note for AP CSA:** When working with `String`, **always** use `.equals()` to compare content.
