<b>Learning objective</b>: Compute statement execution counts & informal run-time comparison of iterative statements

<b>Essential Knowledge</b>: A statement execution count indicates the number of times a statement is executed by the program (e.g., how many times a barista performs a specific task).

<h3> What IS informal code analysis in our Boba Cafe? </h3>

```java
// CODE EXAMPLE #1 (for loop)
// Scenario: We have 30 customers. Every 3rd customer gets a free boba topping!
public class InformalCodeAnalysis {
    public static void main(String[] args) {
        int freeToppings = 0;
        for (int k = 0; k < 30; k++)
        {
            if (k % 3 == 0) // statement 1: Checking if it's a multiple of 3
            {
                freeToppings++; // statement 2: Giving the free topping
            }
        }
        System.out.println("Total free toppings given: " + freeToppings);
    }
}
InformalCodeAnalysis.main(null);
```
