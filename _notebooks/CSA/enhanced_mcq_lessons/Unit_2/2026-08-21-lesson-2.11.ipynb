**nested iteration**
<details><summary>Click to reveal</summary>occurs when we have a loop inside of another loop, similar to nested conditional statements in unit 3. In the cafe, think of this as processing multiple orders in a queue (outer loop), and for each order, adding multiple scoops of toppings (inner loop).</details>

When you have one loop inside another, the inner loop has to finish all its rounds before the outer loop moves to the next round. If the inner loop has a "stop" command, it only stops for that round of the outer loop. The next time the outer loop starts a new round, the inner loop starts over.

If you have two nested loops without stops, and the outer one runs *n* times while the inner one runs *m* times each time the outer one goes around, the inner loop will run *m* times *n* times, which is **m * n times** in total. This rule also applies if you have more than two nested loops. To find the total number of times the innermost loop runs, just multiply how many times each loop runs per round.

**Boba Analogy:** If you have to prepare *n* drink carriers/trays (outer loop), and each tray holds *m* drinks (inner loop), you will make a total of *m * n* drinks. The innermost step (making a drink) runs completely for every tray before moving to the next tray.

```java
public class NestedLoopsDemo {
    public static void main(String[] args) {
        int n = 3; // numb of times the outside loop runs (Trays)
        int m = 2; // numb of times the inside loop runs (Drinks per Tray)

        System.out.println("Starting the afternoon boba rush...\n");
        
        // the nested loops
        for (int i = 1; i <= n; i++) {
            System.out.println("Outer loop iteration: Preparing Tray #" + i);
            for (int j = 1; j <= m; j++) {
                System.out.println("  -> Inner loop iteration: Sealing Drink #" + j + " for Tray #" + i);
            }
            System.out.println("Tray #" + i + " is ready for pickup!\n");
        }
    }
}
NestedLoopsDemo.main(null);
```
