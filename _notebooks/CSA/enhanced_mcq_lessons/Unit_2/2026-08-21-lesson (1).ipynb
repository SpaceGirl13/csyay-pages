## Part 1 — Finding a substring within a string

We can use the **"window" method** to find a smaller string (a substring) inside a larger string.

Think of the "window" like a boba straw sliding across a cup. A window is created matching the exact length of the substring we are looking for. We then iterate through the main string, shifting our window one character at a time, and comparing the window's contents to our target.

### Example: Searching for "OCA" in "TAPIOCA"

**T A P I O C A**

1. `[T A P] I O C A`  (Matches "OCA"? False)
2. `T [A P I] O C A`  (Matches "OCA"? False)
3. `T A [P I O] C A`  (Matches "OCA"? False)
4. `T A P [I O C] A`  (Matches "OCA"? False)
5. `T A P I [O C A]`  (Matches "OCA"? True!)

Let's look at a Java implementation. Run the cell below.

````java
public class StringFinder {
    public static void main(String[] args) {
        String order = "tapioca";
        String sub = "xxx"; // Let's pretend we are searching for an invalid topping
        boolean found = false; // will be set to true once substring is found

        // Iterating forwards: starting at first index (0) and going to the length of the word
        for (int i = 0; i < order.length(); i++) { 
            // 🛑 DANGER ZONE: Look closely at this substring call!
            String portion = order.substring(i, i + sub.length());
            
            if (sub.length() > portion.length()) break;
            
            if (portion.equals(sub)) { // make sure you use .equals!!
                found = true;
                break;
            }
        }

        if (found) {
            System.out.println("Topping found in the order!");
        } else {
            System.out.println("Topping is NOT in the order.");
        }
    }
}

StringFinder.main(null);
````

### 🛑 POPCORN HACK: Run the code... what happened?

If you ran the code above, you probably hit a `StringIndexOutOfBoundsException`. 

**Why did it crash?**
The loop goes all the way to `i = order.length() - 1`. If `order` is "tapioca" (length 7) and `sub` is "xxx" (length 3), on the last loop, `i = 6`. 
`order.substring(6, 6 + 3)` tries to read up to index 9, which doesn't exist! The check `if (sub.length() > portion.length())` happens *after* the crash.

**Another issue:**
What if our substring was the word **"OCA"**? Note that OCA is at the very end of the whole string "TAPIOCA". If our loop stops too early, we might miss it. If it goes too far, we spill the tea (crash).

<details>
<summary><b>How to fix the loop bound (click to reveal)</b></summary>

Change the loop condition to stop exactly when the window reaches the end of the word:
```java
for (int i = 0; i <= order.length() - sub.length(); i++) {
    String portion = order.substring(i, i + sub.length());
    // ...
}
```
</details>
