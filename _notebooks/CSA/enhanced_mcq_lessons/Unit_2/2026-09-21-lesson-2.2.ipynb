## 🍵 2. Comparison Operators

Comparison operators compare two values and produce a Boolean result.

| Operator | Meaning | Boba Example |
|---|---|---|
| `==` | equal to | `toppings == 2` |
| `!=` | not equal to | `price != 5.00` |
| `>` | greater than | `price > 6.00` |
| `<` | less than | `sugar < 50` |
| `>=` | greater than or equal to | `pearls >= 2` |
| `<=` | less than or equal to | `price <= 7.00` |

### 🧠 Remember
`==` **asks** whether two values are equal.

`=` **assigns** a value to a variable.

````java
// Java
double price = 6.50;
int toppings = 2;

System.out.println(price > 5.00);   // true
System.out.println(price < 5.00);   // false
System.out.println(toppings >= 2);  // true
System.out.println(toppings != 1);  // true
````

## 🔍 4. How Do You Evaluate an Expression?

Look at the values first, then apply the operator.

**Example:**

```java
int pearls = 3;
pearls > 2
```

1. `pearls` has the value `3`.
2. The expression becomes `3 > 2`.
3. `3 > 2` is **true**.

So the Boolean expression evaluates to `true`.

## ⚠️ 5. Common Mistake: `=` vs. `==`

This is one of the most important things to know.

```java
int pearls = 3;     // assignment
pearls == 3         // comparison
```

### 🧋 Boba memory trick

**One `=` puts a value INTO the cup.**

**Two `==` asks if the values MATCH.**

## 🍓 6. Strings: An AP CSA Warning

For `String` objects, `==` checks whether two references refer to the same object. It does **not** reliably compare the text inside the strings.

To compare String contents, use `.equals()`.

```java
String drink = "Milk Tea";
drink.equals("Milk Tea")
```

For Unit 2.2, make sure you understand that `==` has different behavior for primitive values versus reference types.
