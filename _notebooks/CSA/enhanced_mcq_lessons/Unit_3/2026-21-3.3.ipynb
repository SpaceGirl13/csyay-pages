# Anatomy of a Boba Cafe Class 🧋

Welcome to the Boba Cafe Management System! Today, we're exploring how to build robust Java classes using Menus, GUIs, and Inheritance, all while managing our delicious boba orders.

## 1. Console-Based Boba Menu
We'll use Java's `Scanner` and `System` classes to handle order input, and a 2D array to manage our delicious toppings.

```java
import java.util.Scanner;
import java.lang.Math;

public class BobaMenu {
    public final String DEFAULT = "\u001B[0m";
    public final String[][] TOPPINGS = {
        {"Boba", "\u001B[35m"}, 
        {"Lychee", "\u001B[36m"}, 
        {"Grass Jelly", "\u001B[32m"}, 
    };
    
    public BobaMenu() {
        Scanner sc = new Scanner(System.in);
        System.out.println("Welcome to Boba Cafe! Select an option:");
        System.out.println("1 - Greeting\n2 - List Toppings\n3 - Start Brewing\n0 - Quit");
        
        boolean quit = false;
        while (!quit) {
            try {
                int choice = sc.nextInt();
                quit = this.action(choice);
            } catch (Exception e) {
                sc.nextLine();
                System.out.println("Invalid input!");
            }
        }
        sc.close();
    }

    private boolean action(int selection) {
        switch (selection) {
            case 1: System.out.println("Welcome to Boba Cafe!"); break;
            case 2: for(String[] t : TOPPINGS) System.out.println(t[1] + t[0] + DEFAULT); break;
            case 3: 
                System.out.print("Brewing...");
                for(int i=0; i<10; i++) {
                    try { Thread.sleep(100); } catch (Exception e) {}
                    System.out.print("🧋");
                }
                System.out.println(" Done!");
                break;
            case 0: return true;
        }
        return false;
    }
    
    static public void main(String[] args) { new BobaMenu(); }
}
BobaMenu.main(null);
```

## 2. Desktop GUI for Boba Ordering
Using `JFrame` to create a visual interface for customers to select their drinks.

```java
import java.awt.event.*;
import javax.swing.*;

public class BobaShopGUI extends JFrame implements ActionListener {
    private JLabel message = new JLabel("Select a drink!");
    
    public BobaShopGUI() {
        JFrame frame = new JFrame("Boba Order System");
        JMenuBar bar = new JMenuBar();
        JMenu menu = new JMenu("Menu");
        
        JMenuItem item1 = new JMenuItem("Milk Tea");
        item1.addActionListener(this);
        menu.add(item1);
        
        bar.add(menu);
        frame.setJMenuBar(bar);
        frame.add(message);
        frame.setSize(300, 200);
        frame.setVisible(true);
    }

    public void actionPerformed(ActionEvent e) {
        message.setText("Ordering: " + e.getActionCommand());
    }

    public static void main(String[] args) { new BobaShopGUI(); }
}
BobaShopGUI.main(null);
```

## 3. Inheritance: The Boba Drink Hierarchy
Inheritance allows us to define a base `BobaDrink` class, and extend it for specific varieties like `MilkTea` and `FruitTea`.

```java
// Superclass
class BobaDrink {
    String flavor = "Base Tea";
    public String toString() { return "A cup of " + flavor; }
}

// Subclass
class MilkTea extends BobaDrink {
    public MilkTea() { this.flavor = "Milk Tea"; }
}

// Testing Inheritance
BobaDrink myDrink = new MilkTea();
System.out.println(myDrink.toString());
```
