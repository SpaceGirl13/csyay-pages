## Core Concept

**Scope**  the region of code where a variable exists and can be used.
**Access modifiers**  keywords that control which *other classes* can see a member.

### Scope Levels (smallest  largest)

| Level | Where declared | Lives until |
|---|---|---|
| **Local variable** | Inside a method or block `{ }` | Method returns |
| **Instance variable** | In a class, outside methods | Object is garbage collected |
| **Class (static) variable** | In a class with `static` | Program ends |

### Access Modifiers

| Modifier | Who can access it |
|---|---|
| `private` | Only code **inside** the same class |
| `public` | Code **anywhere** |

### Encapsulation
The design principle of making instance variables `private` and providing `public` getter/setter methods.
This protects data from invalid changes and hides internal implementation details.

### Key Vocabulary
| Term | Meaning |
|---|---|
| **Scope** | The region of code where a variable is visible |
| **Local variable** | Declared inside a method; inaccessible outside it |
| **Instance variable** | Declared in a class; accessible by all instance methods |
| **Access modifier** | `public` or `private`; controls cross-class visibility |
| **Encapsulation** | Hiding fields with `private`; exposing through `public` methods |
| **Getter (accessor)** | `public` method that returns a `private` field |
| **Setter (mutator)** | `public` method that modifies a `private` field (often with validation) |


````java
// 
// MenuItem: full encapsulation example
// 
public class MenuItem {

    // PRIVATE instance variables  only accessible inside this class
    private String  name;
    private double  price;
    private boolean isVegan;
    private int     sugarLevel;   // 0100

    // Constructor
    public MenuItem(String name, double price, boolean isVegan, int sugarLevel) {
        this.name       = name;
        this.price      = price;
        this.isVegan    = isVegan;
        this.sugarLevel = sugarLevel;
    }

    //  Getters (accessors) 
    public String  getName()       { return name; }
    public double  getPrice()      { return price; }
    public boolean isVegan()       { return isVegan; }
    public int     getSugarLevel() { return sugarLevel; }

    //  Setters (mutators) with validation 
    public void setPrice(double price) {
        if (price >= 0) {
            this.price = price;   // only update if valid
        }
        // silently ignores negative prices
    }

    public void setSugarLevel(int level) {
        if (level >= 0 && level <= 100) {
            this.sugarLevel = level;
        }
    }

    //  Method showing LOCAL variable scope 
    public String describe() {
        String tag = isVegan ? " Vegan" : "Non-vegan";  // LOCAL  only lives here
        String sugar = sugarLevel + "% sugar";             // LOCAL  only lives here
        return name + " | " + tag + " | " + sugar
             + " | $" + String.format("%.2f", price);
        // tag and sugar go out of scope here
    }
}

System.out.println("MenuItem class ready ");
````

## Part 1  Private Fields and Encapsulation


````java
MenuItem boba = new MenuItem("Brown Sugar Boba", 5.75, false, 75);
System.out.println(boba.describe());

//  Trying to access a private field directly 
// boba.price = 3.00;         COMPILE ERROR: price has private access
// boba.sugarLevel = 200;     COMPILE ERROR: private

//  The correct way: use public methods 
boba.setPrice(6.25);
System.out.println("Updated price: $" + boba.getPrice());

boba.setSugarLevel(200);                      // rejected  out of range
System.out.println("Sugar (unchanged): " + boba.getSugarLevel() + "%");

boba.setSugarLevel(50);                       // accepted
System.out.println("Sugar (updated):   " + boba.getSugarLevel() + "%");
````

## Part 2  Local Variable Scope


````java
public class CafeRegister {
    private double dailyRevenue = 0.0;     // INSTANCE variable  persists

    public void processOrder(double subtotal, double taxRate) {
        double tax   = subtotal * taxRate;  // LOCAL variable
        double total = subtotal + tax;      // LOCAL variable

        dailyRevenue += total;              // updates instance variable

        System.out.println("Order total: $" + String.format("%.2f", total));
        // 'tax' and 'total' are out of scope after this method returns
    }

    public double getDailyRevenue() {
        // System.out.println(total);   COMPILE ERROR: 'total' is out of scope here
        return dailyRevenue;
    }
}

CafeRegister register = new CafeRegister();
register.processOrder(12.50, 0.0975);
register.processOrder( 8.75, 0.0975);
System.out.println("Daily revenue: $" + String.format("%.2f", register.getDailyRevenue()));
````

## Part 3  Variable Shadowing (Common MCQ Trap)

When a local variable has the **same name** as an instance variable,
the local variable **shadows** (hides) the instance variable within that scope.


````java
public class DrinkStation {
    private int stockCount = 100;     // instance variable

    public void addStock(int stockCount) {
        // Inside this method, 'stockCount' refers to the PARAMETER
        // The instance variable is shadowed
        System.out.println("Adding: " + stockCount);       // prints the parameter
        System.out.println("Before: " + this.stockCount);  // 'this.' reaches the field
        this.stockCount += stockCount;
        System.out.println("After:  " + this.stockCount);
    }
}

DrinkStation station = new DrinkStation();
station.addStock(50);
````

## Vocabulary Recap

| Term | Definition |
|---|---|
| **Scope** | The region of code where a variable is visible and usable |
| **Local variable** | Declared inside a method; only exists while that method runs |
| **Instance variable** | Declared in the class body; persists for the lifetime of the object |
| **Access modifier** | `public` or `private`; controls cross-class visibility |
| **`private`** | Only accessible inside the same class |
| **`public`** | Accessible from any class |
| **Encapsulation** | Hiding data behind `private` fields + `public` accessor methods |
| **Getter (accessor)** | Returns a private field |
| **Setter (mutator)** | Modifies a private field, often with validation |
| **Variable shadowing** | A local variable hides an instance variable with the same name |
