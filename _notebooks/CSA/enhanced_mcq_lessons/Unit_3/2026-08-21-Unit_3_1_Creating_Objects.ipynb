# AP CSA Unit 3.1 — Creating Objects 🧋

Welcome to the Boba Cafe! Unit 3.1 marks the beginning of our journey into **Class Creation**. This unit focuses on creating objects from classes and understanding how objects store information and perform actions.

## 🎯 Learning Objectives
- Understand what a class is.
- Understand what an object is.
- Create objects using the `new` keyword.
- Use constructors to initialize objects.
- Store objects in variables.
- Call methods on objects.

## 1. Classes vs. Objects 🧋
- **A class** is like a blueprint for creating objects.
- **An object** is an actual instance created from that class.

### Think of a boba shop:
- **Class:** `BobaDrink`
- **Objects:** Mango Boba, Taro Boba, Matcha Boba

The class describes what a boba drink can have and do, while each object represents a specific drink.

## 2. Creating an Object 🧋
In Java, objects are usually created using the `new` keyword.

````java
// Example: Creating a basic boba drink object
// Note: Assuming a BobaDrink class is defined in your project
// BobaDrink drink = new BobaDrink();
````

### Anatomy of Object Creation:
```java
BobaDrink drink = new BobaDrink();
```
- `BobaDrink` ↓ data type
- `drink` ↓ variable/reference
- `new BobaDrink()` ↓ creates the object

The variable `drink` now refers to a `BobaDrink` object.

## 3. Constructors 🧋
A constructor is used when an object is created to initialize its starting values.

````java
// Example: Using a parameterized constructor
// BobaDrink drink = new BobaDrink("Mango", "Boba");
// "Mango" → flavor
// "Boba"  → topping
````

## 4. Object References 🧋
When you create an object, the variable stores a **reference** (a memory address pointing to the object), not the raw data itself.

```java
String drink = new String("Mango Boba");
```
`drink` doesn't directly contain the entire object the way an `int` contains a number. Instead, think of it like:
`drink` ───────→ `"Mango Boba"` object

## 5. Calling Methods on Objects 🧋
Objects can have methods, which are actions the object can perform. You call a method using the **dot operator (`.`)**:

```java
object.method();
```

````java
public class MethodCallDemo {
    public static void main(String[] args) {
        String drink = "Mango Boba";
        
        System.out.println("Length: " + drink.length());
        System.out.println("Uppercase: " + drink.toUpperCase());
    }
}
MethodCallDemo.main(null);
````

## 6. Using Objects with Variables 🧋
You can create multiple objects from the same class. Each can hold completely different state information.

````java
// Multiple objects from the same class blueprint:
// BobaDrink drink1 = new BobaDrink("Mango");
// BobaDrink drink2 = new BobaDrink("Taro");
// drink1 points to the Mango object, drink2 points to the Taro object.
````

## 7. Object State & Behavior 🧋
- **Object State:** The information currently stored in that object (e.g., flavor, size, topping, sugar level, ice level).
- **Object Behavior:** What the object can do via methods (e.g., `addTopping()`, `makeDrink()`, `getPrice()`).

## ⭐ AP CSA Cheat Sheet

| Term | Meaning |
| :--- | :--- |
| **Class** | Blueprint for creating objects |
| **Object** | An instance of a class |
| **`new`** | Creates an object |
| **Constructor** | Initializes a new object |
| **Reference** | A variable that refers to an object |
| **Method** | Behavior/action associated with an object |
| **State** | Information stored in an object |
| **Dot operator (`.`)** | Used to access methods/fields |
