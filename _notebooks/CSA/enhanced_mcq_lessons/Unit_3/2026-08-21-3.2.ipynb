# AP CSA Unit 3.2 — Creating and Using Objects 🧋

Welcome back to the Boba Cafe! Unit 3.2 focuses on creating and using objects, including constructors, methods, parameters, and object references.

## 🎯 Learning Objectives
- Create objects from classes.
- Understand and use constructors.
- Pass arguments to constructors.
- Call methods on objects.
- Pass arguments to methods.
- Understand object references.

## 1. Creating Objects 🧋
A class is a blueprint, while an object is an instance of that class.
Think: **Class = recipe**, **Object = actual boba drink**.

````java
// Example: Creating a BobaDrink object
// BobaDrink drink = new BobaDrink();
// Breakdown:
// BobaDrink -> the class/data type
// drink     -> the reference variable
// new BobaDrink() -> creates the object
````

## 2. Constructors 🧋
A constructor is used to initialize an object when it is created. It is called automatically when you use `new`.

**General Pattern:**
```java
ClassName objectName = new ClassName(arguments);
```

````java
// Example: Passing arguments to a constructor
// BobaDrink drink = new BobaDrink("Mango", "Boba");
// "Mango" -> flavor
// "Boba"  -> topping
````

## 3. Parameters vs. Arguments 🧋
- **Parameter**: A variable listed in a method or constructor definition (a placeholder).
  - Example: `public BobaDrink(String flavor, String topping)` -> `flavor` and `topping` are parameters.
- **Argument**: The actual value passed to the constructor or method.
  - Example: `new BobaDrink("Mango", "Boba");` -> `"Mango"` and `"Boba"` are arguments.

🧠 **Remember:** Parameter = placeholder | Argument = actual value

## 4. Calling Methods & Methods with Parameters 🧋
Objects have methods that allow you to perform actions or get information using the dot operator (`object.method();`). Some methods also require arguments.

````java
public class BobaMethodDemo {
    public static void main(String[] args) {
        String drink = "Mango Boba";
        int length = drink.length(); // Calling a method that returns a value
        System.out.println("Drink length: " + length);
        
        // Example with parameter/argument:
        // drink.addTopping("Jelly");
    }
}
BobaMethodDemo.main(null);
````

## 5. Methods That Return Values vs. Void Methods 🧋
- **Returning Values**: Methods like `drink.length()` or `drink.getPrice()` produce and return information that you can store or print.
- **Void Methods**: Methods with return type `void` (like `drink.addTopping("Boba");`) perform an action or change the object's state, but do not give a value back.

## 6. Object References & Multiple References 🧋
When you create an object, a reference variable points to that object in memory.

### Multiple References:
Two variables can refer to the same object:
```java
BobaDrink drink1 = new BobaDrink("Taro");
BobaDrink drink2 = drink1;
```
If the object is changed through one reference, the change can be seen through the other reference!

## ⭐ AP CSA Cheat Sheet

| Term | Meaning |
| :--- | :--- |
| **Class** | Blueprint for objects |
| **Object** | Instance of a class |
| **Constructor** | Initializes an object |
| **Parameter** | Variable in a method/constructor definition |
| **Argument** | Actual value passed to a method/constructor |
| **Method** | Performs an action or returns information |
| **Return value** | Information produced by a method |
| **`void`** | Method doesn't return a value |
| **Reference** | Variable that points to an object |
| **`new`** | Creates an object |
