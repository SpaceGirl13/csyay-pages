# AP CSA Unit 3.5 — Method Overrides & The Super Keyword 🧋

Welcome back to the Boba Cafe! In this lesson, we explore how subclasses inherit methods from their parent class and how we can customize them using **Method Overrides** and the **`super` keyword**.

## 1. Method Overrides 🧋

Every class inherits methods from its **parent class** [cite: 25]. These methods can be called directly in the child class and can also be **overridden** to provide specialized behavior [cite: 25].

## 2. BobaContainer Base Class 🧋

Imagine our base class representing standard rectangular boba packaging containers.

```java
public class BobaContainer {
    protected String name;
    private int length;
    private int width;

    // Default constructor
    public BobaContainer() {
        this.name = "container";
        this.length = 10;
        this.width = 5;
    }

    // Parameterized constructor
    public BobaContainer(String name, int length, int width) {
        this.name = name;
        this.length = length;
        this.width = width;
    }

    // Getter methods
    public String get_name() {
        return this.name;
    }

    public int get_length() {
        return this.length;
    }

    public int get_width() {
        return this.width;
    }

    // Setter methods
    public void set_name(String n) {
        this.name = n;
    }

    public void set_length(int a) {
        this.length = a;
    }

    public void set_width(int b) {
        this.width = b;
    }

    // Method to calculate the area
    public double calc_area() {
        return this.length * this.width;
    }

    // Method to calculate the perimeter
    public double calc_perimeter() {
        return 2 * this.length + 2 * this.width;
    }

    // Method to print container info
    public void print_shape() {
        System.out.println("Container: " + this.name);
    }

    public void print_something() {
        System.out.println("This is a standard boba container");
    }
}

public class Driver {
    public static void main(String[] args) {
        BobaContainer c1 = new BobaContainer();
    }
}
```

### Creating a Triangle Boba Box:

Our base `BobaContainer` only takes length and width parameters, which works fine for rectangular boxes. But what if our specialty boba shop introduces a **Triangle Boba Box**? 

To define a triangle box, we can use 3 side lengths while still inheriting behavior from the base `BobaContainer` class [cite: 25].

```java
public class TriangleBobaBox extends BobaContainer {
    private int side1;
    private int side2;
    private int side3;

    public TriangleBobaBox() {
        this.name = "triangle box";
        this.side1 = 1;
        this.side2 = 2;
        this.side3 = 3;
    }

    // Constructor that takes a name and three side lengths
    public TriangleBobaBox(String name, int s1, int s2, int s3) {
        super(name, 0, 0); // Call parent constructor to set name
        this.name = "triangle box";
        this.side1 = s1;
        this.side2 = s2;
        this.side3 = s3;
    }
}
```

## 3. We Have a Problem... 🧋

We inherited the default container's `calc_area()` method [cite: 25]:
$$\text{area} = \text{length} \times \text{width}$$
This defaults to an area of `0` when we create a triangle box without specifying length and width [cite: 25]. 

Instead, we can use **Heron's formula** to calculate the area of a triangle given 3 side lengths [cite: 25]:
$$\text{area} = \sqrt{s(s-s_1)(s-s_2)(s-s_3)}$$
where $s$ is the semi-perimeter: $s = \frac{s_1 + s_2 + s_3}{2}$ [cite: 25].

The **`@Override`** annotation tells Java that the following method is an override [cite: 25]. If the method does not correctly match an existing parent method, the compiler throws an error [cite: 25]. This is excellent practice for catching spelling mistakes and ensuring proper method overriding [cite: 25].

## Important Rules to Note 🧋
1. Using the **`final`** keyword in a parent method makes that method **unable** to be overridden [cite: 25].
2. Methods can be overridden to give *more* access (e.g., `protected` $\rightarrow$ `public`), but cannot restrict access (e.g., `public` $\rightarrow$ `private`) [cite: 25].
3. Use of **`@Override`** is highly encouraged [cite: 25].

## 4. The `super` Keyword 🧋

What if we want to create our child triangle box class, but also invoke the parent container's version of an overridden method?

For example, what if we want to print out both that our item is a standard boba container *and* that it is a triangle box? We can accomplish this using the **`super` keyword** [cite: 25].

```java
public class TriangleBobaBox extends BobaContainer {
    private int side1;
    private int side2;
    private int side3;

    public TriangleBobaBox(String name, int s1, int s2, int s3) {
        super(name, 0, 0);
        this.name = "triangle box";
        this.side1 = s1;
        this.side2 = s2;
        this.side3 = s3;
    }

    public void print_shape() {
        super.print_something(); // Calls parent method first
        print_something();       // Calls subclass method
        System.out.println("Container: " + this.name);
    }

    @Override
    public void print_something() {
        System.out.println("This is a triangle boba box");
    }
}

TriangleBobaBox t2 = new TriangleBobaBox("triangle box", 3, 4, 5);
t2.print_shape();
```

## Summary of `super` 🧋

* Using `super.method()` calls the parent class's method [cite: 25]. Calling a super method inside another method will first complete the parent method before moving on to the next line [cite: 25].
* **Warning:** Be careful not to call a method on itself without the `super` keyword inside an override, as this will cause an **infinite recursion** error [cite: 25]!
