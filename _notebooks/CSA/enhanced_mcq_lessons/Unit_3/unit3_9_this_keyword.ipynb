## Core Concept

`this` is a reference to the **current object**  the specific instance on which a method or constructor is being called.

### Two Main Uses

**1  Resolve parameter shadowing in constructors**

When a constructor parameter has the **same name** as an instance variable, `this.` tells Java you mean the field:

```java
public MenuItem(String name, double price) {
    this.name  = name;   // this.name  = instance variable (LEFT)
    this.price = price;  // price      = parameter         (RIGHT)
}
```

**2  Delegate to another constructor with `this(...)`**

Calls a different constructor in the *same class*. Must be the **very first line**.

```java
public MenuItem(String name, double price) {
    this(name, price, "Medium");   // calls the 3-arg constructor
}
```

### Key Vocabulary
| Term | Meaning |
|---|---|
| **`this`** | Reference to the current object instance |
| **Parameter shadowing** | A parameter name matches an instance variable name; `this.` distinguishes them |
| **Constructor delegation** | Using `this()` to forward a constructor call to another constructor in the same class |
| **Implicit parameter** | The object `this` refers to; passed automatically with every instance method call |


````java
// 
// Use 1  Resolving parameter shadowing
// 
public class MenuItem {
    private String name;
    private double price;
    private String size;

    // Without 'this.', the assignment name = name does NOTHING useful
    // (assigns parameter to itself  the instance variable stays at default)
    public MenuItem(String name, double price, String size) {
        this.name  = name;    // LEFT  = instance variable
        this.price = price;   // RIGHT = parameter value
        this.size  = size;
    }

    public String toString() {
        return "[" + size + "] " + name + "  $" + String.format("%.2f", price);
    }
}

MenuItem m = new MenuItem("Taro Boba", 6.50, "Large");
System.out.println(m);
````

## Part 2  Constructor Delegation with `this(...)`

Use `this()` to avoid repeating initialization logic across multiple constructors.
The called constructor must exist, and `this()` must be the **first** statement.


````java
// 
// Constructor delegation chain
// 
public class DrinkOrder {
    private String drinkName;
    private String size;
    private int    sugarLevel;   // 0, 25, 50, 75, or 100
    private boolean addBoba;

    //  Most specific constructor (all args) 
    public DrinkOrder(String drinkName, String size, int sugarLevel, boolean addBoba) {
        this.drinkName  = drinkName;
        this.size       = size;
        this.sugarLevel = sugarLevel;
        this.addBoba    = addBoba;
    }

    //  3-arg: defaults addBoba to false 
    public DrinkOrder(String drinkName, String size, int sugarLevel) {
        this(drinkName, size, sugarLevel, false);   // delegates up 
    }

    //  2-arg: defaults sugar to 100%, no boba 
    public DrinkOrder(String drinkName, String size) {
        this(drinkName, size, 100);                 // delegates up 
    }

    //  1-arg: defaults to Medium, 100% sugar, no boba 
    public DrinkOrder(String drinkName) {
        this(drinkName, "Medium");                  // delegates up 
    }

    public String toString() {
        return "[" + size + "] " + drinkName
             + " | Sugar: " + sugarLevel + "%"
             + " | Boba: " + (addBoba ? "yes" : "no");
    }
}

//  Demo  all four constructors 
DrinkOrder d1 = new DrinkOrder("Matcha Latte", "Large", 75, true);
DrinkOrder d2 = new DrinkOrder("Oolong Tea",   "Small", 50);
DrinkOrder d3 = new DrinkOrder("Jasmine Tea",  "Medium");
DrinkOrder d4 = new DrinkOrder("Brown Sugar Boba");

System.out.println(d1);
System.out.println(d2);
System.out.println(d3);
System.out.println(d4);
````

## Part 3  `this` in Instance Methods

`this` can also appear in instance methods  useful for returning the current object
or passing it to another method.


````java
public class LoyaltyCard {
    private String ownerName;
    private int    points;

    public LoyaltyCard(String ownerName) {
        this.ownerName = ownerName;
        this.points    = 0;
    }

    // Returns 'this' so calls can be chained
    public LoyaltyCard addPoints(int amount) {
        this.points += amount;
        return this;                  // returns the current object
    }

    public void printStatus() {
        System.out.println(ownerName + " has " + points + " points");
    }
}

//  Method chaining (returns this each time) 
LoyaltyCard card = new LoyaltyCard("Anishka");
card.addPoints(10)
    .addPoints(25)
    .addPoints(15)
    .printStatus();
````

## Common Mistakes

| Mistake | What happens | Fix |
|---|---|---|
| `name = name;` in constructor | Assigns parameter to itself; field stays `null`/0 | Use `this.name = name;` |
| `this()` not on first line | Compile error | Move it to line 1 of the constructor |
| Calling `this()` in a regular method | Compile error  only legal in constructors | Use a regular method call instead |
| Infinite delegation loop: `A()` calls `this()` which calls `A()` | Compile error | Ensure at least one constructor doesn't delegate |

| Term | Definition |
|---|---|
