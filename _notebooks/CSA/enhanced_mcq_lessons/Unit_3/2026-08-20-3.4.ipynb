# AP CSA Unit 3.4 — Constructors 🧋

According to the current College Board AP CSA framework, Unit 3.4 focuses on constructors and initializing the state of objects in our Boba Cafe.

## 🎯 Learning Objectives
- Understand an object's state.
- Declare instance variables.
- Write constructors.
- Use constructor parameters to initialize instance variables.
- Understand the default constructor and Java's default values.
- Understand why copying mutable object parameters can matter.

## 1. Object State 🧋
An object's **state** is the collection of its attributes and their current values.
For example, a `BobaDrink` object might have fields like `flavor`, `sugar`, and `hasPearls`.

```java
// Example: Object state representation in Java
public class BobaStateDemo {
    private String flavor = "Mango";
    private int sugar = 50;
    private boolean hasPearls = true;
}
```

## 2. Instance Variables 🧋
An **instance variable** is a variable that belongs to an object instance.
Each object gets its own separate copy of these variables, meaning multiple drinks can have completely independent states.

```java
public class BobaDrink {
    private String flavor;
    private int sugar;
    private boolean hasPearls;
}
```

## 3. Constructors Initialize State & Parameters 🧋
A **constructor** is used to set the initial state of an object when it's created using `new`.
- **Parameter**: A variable listed in a constructor definition (placeholder).
- **Argument**: The actual value passed when calling the constructor.

```java
public class BobaConstructorDemo {
    private String flavor;
    private int sugar;

    // Constructor with parameters
    public BobaConstructorDemo(String flavor, int sugar) {
        this.flavor = flavor;
        this.sugar = sugar;
    }

    public static void main(String[] args) {
        // "Matcha" and 50 are arguments passed to the constructor
        BobaConstructorDemo drink = new BobaConstructorDemo("Matcha", 50);
        System.out.println("Drink created successfully!");
    }
}
BobaConstructorDemo.main(null);
```

## 4. The `this` Keyword 🧋
When a constructor parameter has the same name as an instance variable, Java uses `this` to distinguish between them:
- `this.flavor` refers to the **instance variable**.
- `flavor` refers to the **parameter**.

## 5. Default Constructor & Java's Default Values 🧋
If you write **no constructor at all**, Java provides a default no-parameter constructor and initializes instance variables to default values:

| Data Type | Default Value |
| :--- | :--- |
| `int` | `0` |
| `double` | `0.0` |
| `boolean` | `false` |
| Reference types (e.g., `String`) | `null` |

⚠️ **Important Rule:** If you write *any* custom constructor with parameters, Java **no longer** automatically provides the default no-parameter constructor. You must write it explicitly if you want both!

## 6. Mutable Object Parameters 🧋
When a constructor receives a mutable object as a parameter, simply assigning `this.someObject = someObject;` links them to the same memory reference. Making a **copy** of the mutable object prevents external code from accidentally mutating your object's internal state.

## 🧋 Quick Practice
- **Question 1:** What is the purpose of a constructor?
  * *Answer:* To initialize the state of a newly created object.
- **Question 2:** What is the default value of an `int` instance variable?
  * *Answer:* `0`
- **Question 3:** What does `this.flavor` refer to?
  * *Answer:* The flavor instance variable belonging to the current object.
- **Question 4:** What happens if you write a constructor with parameters?
  * *Answer:* Java no longer automatically provides the no-parameter constructor.
