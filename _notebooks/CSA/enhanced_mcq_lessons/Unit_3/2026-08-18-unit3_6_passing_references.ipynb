## Core Concept

When you pass an **object** to a method, Java passes its **reference**  a pointer to the same object in memory.
Changes made to the object inside the method **affect the original**.

Primitives (`int`, `double`, `boolean`) are passed by **value**  the method gets a copy, so the original never changes.

| Argument type | What gets passed | Original affected? |
|---|---|---|
| `int`, `double`, `boolean` | A copy of the value | No No |
| Any object (`MenuItem`, `String`, ) | A copy of the **reference** (same object) | <- CORRECT Yes |

### Key Vocabulary
| Term | Meaning |
|---|---|
| **Reference** | A variable that holds the memory address of an object |
| **Pass by value** | Java always passes a *copy*  of the value for primitives, of the address for objects |
| **Object reference parameter** | A method parameter whose type is a class; both the caller and method point to the same object |
| **Return type** | The data type declared before a method name; can be a class type |
| **`null`** | A reference that points to no object |


````java
// 
// Setup: MenuItem class  used throughout this lesson
// 
public class MenuItem {
    private String name;
    private double price;

    public MenuItem(String name, double price) {
        this.name  = name;
        this.price = price;
    }

    public String getName()         { return name; }
    public double getPrice()        { return price; }
    public void   setPrice(double p){ price = p; }

    @Override
    public String toString() {
        return name + "  $" + String.format("%.2f", price);
    }
}

System.out.println("MenuItem class ready ");
````

## Part 1  Passing an Object Reference


````java
// A method that receives a MenuItem REFERENCE and modifies it
public static void applyDiscount(MenuItem item, double percent) {
    // 'item' points to the SAME object the caller passed
    double discounted = item.getPrice() * (1 - percent);
    item.setPrice(discounted);        //  changes the original object
}

//  Demo 
MenuItem matcha = new MenuItem("Matcha Latte", 6.00);
System.out.println("Before discount: " + matcha);

applyDiscount(matcha, 0.10);         // 10% off
System.out.println("After  discount: " + matcha);
// matcha's price IS changed  we modified the object through its reference
````

````java
//  Contrast: primitives are NOT changed 
public static void tryToDouble(int x) {
    x = x * 2;    // only changes the LOCAL copy of x
}

int sugarLevel = 50;
tryToDouble(sugarLevel);
System.out.println("sugarLevel after tryToDouble: " + sugarLevel);
// Still 50  the method only changed its own copy
````

## Part 2  Two References, One Object

Multiple variables can point to the **same** object. Modifying it through one reference changes what you see through the other.


````java
//  Two references, one object 
MenuItem original = new MenuItem("Taro Boba", 5.50);
MenuItem alias    = original;        // both point to the SAME object

alias.setPrice(4.00);                // modifying via 'alias'...
System.out.println(original);        // ...is visible through 'original'

// Contrast: reassigning 'alias' does NOT affect 'original'
alias = new MenuItem("Oolong Tea", 4.50);   // alias now points to a NEW object
System.out.println("original still: " + original);
System.out.println("alias now:      " + alias);
````

## Part 3  Returning an Object


````java
// Define the class first
class MenuItem {
    private String name;
    private double price;
    
    public MenuItem(String name, double price) {
        this.name = name;
        this.price = price;
    }
    
    public double getPrice() { return price; }
    public void setPrice(double price) { this.price = price; }
    public String toString() { return name + " ($" + price + ")"; }
}

// Now the methods and demo work
MenuItem special = new MenuItem("Strawberry Matcha", 6.50 * 0.80);
System.out.println("Today's special: " + special);

MenuItem a = new MenuItem("Jasmine Tea", 3.50);
MenuItem b = new MenuItem("Boba Classic", 5.00);
MenuItem cheapest = (a.getPrice() <= b.getPrice()) ? a : b;
System.out.println("Cheapest: " + cheapest);
````

**Q2.** What does this print?
```java
public static void reset(MenuItem m) {
    m = new MenuItem("Water", 0.00);
}
MenuItem drink = new MenuItem("Boba", 5.00);
reset(drink);
System.out.println(drink.getPrice());
```
- A) 0.0
- B) 5.0 <- CORRECT
- C) null
- D) Runtime error

> **Why B?** `reset` reassigns its local parameter `m` to a new object, but `drink` in the caller still points to the original Boba object. Reassigning the parameter never changes what the caller's variable points to.

## Vocabulary Recap

| Term | Definition |
|---|---|
| **Reference** | Variable storing the memory address of an object |
| **Pass by value** | Java passes a copy  primitives copy the value, objects copy the address |
| **Object reference parameter** | Method receives the same object's address as the caller |
| **Return type** | Data type before the method name; `void` means nothing returned |
| **`null`** | Reference pointing to no object; calling methods on it  `NullPointerException` |
