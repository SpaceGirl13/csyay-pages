## Core Concept

Most members of a class belong to **individual objects** (instance members).
Add `static` and the member belongs to the **class itself**  one shared copy for all objects.

| | Instance | Class (`static`) |
|---|---|---|
| **Variable** | Each object gets its own copy | One shared copy across all objects |
| **Method** | Called on an object: `myOrder.getTotal()` | Called on the class: `CafeOrder.getTotalOrders()` |
| **Created when** | A new object is instantiated | The class is first loaded |
| **`this` available?** | <- CORRECT Yes | No No  no specific object |

### Key Vocabulary
| Term | Meaning |
|---|---|
| **Instance variable / method** | Belongs to one object; each object has its own copy |
| **Class variable** | `static` field; one copy shared by all instances |
| **Class method** | `static` method; called on the class, not an object |
| **`static`** | Keyword marking a member as belonging to the class |
| **`final`** | Keyword making a variable a constant (cannot be reassigned) |


```java
// 
// CafeOrder: shows static vs. instance members side by side
// 
public class CafeOrder {

    //  Class variables (static) 
    private static int    totalOrders = 0;         // shared counter
    private static final double TAX_RATE = 0.0975; // constant, never changes

    //  Instance variables 
    private String customerName;
    private double subtotal;
    private int    orderNumber;

    //  Constructor 
    public CafeOrder(String customerName, double subtotal) {
        this.customerName = customerName;
        this.subtotal     = subtotal;
        totalOrders++;                   // update the SHARED counter
        this.orderNumber = totalOrders;  // stamp this order's unique number
    }

    //  Instance methods 
    public double getTotal() {
        return subtotal * (1 + TAX_RATE);
    }
    public String getReceipt() {
        return "Order #" + orderNumber + " | " + customerName
             + " | Subtotal: $" + String.format("%.2f", subtotal)
             + " | Total: $"    + String.format("%.2f", getTotal());
    }

    //  Class (static) methods 
    public static int    getTotalOrders() { return totalOrders; }
    public static double getTaxRate()     { return TAX_RATE; }
}

System.out.println("CafeOrder class ready ");
```

## Part 1  Using Static vs. Instance Members


```java
//  Create a few orders 
CafeOrder o1 = new CafeOrder("Anishka", 12.50);
CafeOrder o2 = new CafeOrder("Maya",     8.75);
CafeOrder o3 = new CafeOrder("Priya",   15.00);

// Instance methods  called on a specific object
System.out.println(o1.getReceipt());
System.out.println(o2.getReceipt());
System.out.println(o3.getReceipt());

System.out.println();

// Static methods  called on the CLASS, not on any one object
System.out.println("Total orders placed: " + CafeOrder.getTotalOrders());
System.out.println("Tax rate:            " + CafeOrder.getTaxRate());
```

## Part 2  Why Static? (The Shared Counter)

Every time a `CafeOrder` is created, `totalOrders` increments.
If `totalOrders` were an **instance variable**, each object would have its own counter stuck at 1.
Making it `static` means all objects share and update the same number.


```java
//  Visualizing the shared counter 
// (Re-run from scratch  class needs to be redefined to reset counter)
public class Ticket {
    private static int nextId = 1;       // class variable: shared
    private int id;                      // instance variable: unique per ticket

    public Ticket() {
        this.id = nextId;
        nextId++;
    }

    public int getId()             { return id; }
    public static int getNextId()  { return nextId; }
}

Ticket t1 = new Ticket();
Ticket t2 = new Ticket();
Ticket t3 = new Ticket();

System.out.println("t1 id: " + t1.getId());   // 1
System.out.println("t2 id: " + t2.getId());   // 2
System.out.println("t3 id: " + t3.getId());   // 3
System.out.println("Next available id: " + Ticket.getNextId());  // 4
```

## Part 3  Static Methods Cannot Use `this`

A static method has no object context. It cannot access instance variables directly.


```java
public class MenuHelper {
    private String note = "Remember to smile!";   // instance variable

    // <- CORRECT Instance method  has access to 'this'
    public void printNote() {
        System.out.println(this.note);
    }

    // <- CORRECT Static method  works only with its own parameters or other static members
    public static double applyTax(double price, double taxRate) {
        return price * (1 + taxRate);
    }

    // No This would be a COMPILE ERROR:
    // public static void badMethod() {
    //     System.out.println(this.note);  // 'this' doesn't exist in static context
    // }
}

// Calling the static method  no object needed
double total = MenuHelper.applyTax(5.75, 0.0975);
System.out.println("Total with tax: $" + String.format("%.2f", total));
```

## Vocabulary Recap

| Term | Definition |
|---|---|
| **Instance variable** | Non-static field; each object has its own copy |
| **Class variable** | `static` field; one copy shared by all instances |
| **Instance method** | Non-static method; called on an object; can use `this` |
| **Class method** | `static` method; called on the class; no `this` |
| **`static`** | Declares a member as belonging to the class, not individual objects |
| **`final`** | Prevents a variable from being reassigned after initialization |
